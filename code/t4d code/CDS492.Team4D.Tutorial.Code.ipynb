### Stage 1: Regime Context

Intent:
- build a station-level baseline from the early valid years at each station
- turn that baseline into a small, interpretable regime label used later by stage 5 and stage 9

Result to expect:
- `station_baseline` and `station_baseline_display`, both carrying the assigned regime cluster
- clustering metadata saved in memory for the final model bundle


```python
# station character baseline (first valid years, not fixed 1995-2000)
# this keeps the idea simple: each station gets its own early baseline window

# leave water intact for now...
water_base = water.copy( )
water_base = water_base.loc[ 
    :, [ 
        'id',
        'region', 
        'station',
        'datetime',
        'water_temp',
        'salinity',
        'oxygen',
        'oxy_saturation',
        'ph',
        'depth'
    ]
].dropna( subset = [ 'datetime' ] )
water_base[ 'year' ] = water_base[ 'datetime' ].dt.year
#water_base.describe( ).T
```

```python
# annual coverage check so thin years do not define the baseline
year_obs = ( 
    water_base
    .groupby( [ 'region', 'station', 'year' ], as_index = False )
    .agg( n_obs = ( 'datetime', 'size' ) )
)

# leap years and all that jazz
# prolly not a big diff, though
year_obs[ 'expected_obs' ] = np.where( 
    year_obs[ 'year' ] % 4 == 0,
    366 * 24,
    365 * 24,
)

#year_obs.describe( ).T
```

```python
# try to identify years with enough coverage to be considered valid for defining the baseline at all
year_obs[ 'coverage_frac' ] = year_obs[ 'n_obs' ] / year_obs[ 'expected_obs' ]
year_obs[ 'year_is_valid' ] = year_obs[ 'coverage_frac' ] >= 0.70 

#year_obs.describe( ).T
#year_obs[ 'year_is_valid' ].value_counts( )
```

```python
valid_years = year_obs.loc[ year_obs[ 'year_is_valid' ] ].copy( )
valid_years = valid_years.sort_values( [ 'region', 'station', 'year' ] ).reset_index( drop = True )
valid_years[ 'valid_year_rank' ] = valid_years.groupby( [ 'region', 'station' ] ).cumcount( ) + 1
#valid_years.sample( 10 )
del( year_obs ) # done with it
```

```python
# turned out too many didn't even start operating between 95 and 00
# so had to rework to find their first five instead
station_baseline_years = valid_years.loc[ valid_years[ 'valid_year_rank' ] <= 5 ].copy( )

station_baseline_window = ( 
    station_baseline_years
    .groupby( [ 'region', 'station' ], as_index = False )
    .agg( 
        baseline_start_year = ( 'year', 'min' ),
        baseline_end_year = ( 'year', 'max' ),
        n_valid_years = ( 'year', 'size' ),
        mean_year_coverage = ( 'coverage_frac', 'mean' )
    )
)

# is there enough?
station_baseline_window[ 'insufficient_baseline' ] = station_baseline_window[ 'n_valid_years' ] < 5
#baseline_window.describe( ).T
#baseline_window.sample( 10 )
```

```python
# now we can merge that baseline window info back to the base data, 
# and flag which rows are in the baseline window for each station
# so here, a func


def build_season_signature( group ):
    group = group.sort_values( 'month' ).copy( )

    temp_series = group[ 'water_temp_mmean' ]
    oxygen_series = group[ 'oxygen_mmean' ]

    # if there are any valid temp values, we can define warm and cold months based on the mean
    if temp_series.notna( ).any( ):
        temp_mean = float( temp_series.mean( ) )
        warm_mask = ( temp_series > temp_mean ).fillna( False )
        cold_mask = ( temp_series < temp_mean ).fillna( False )

        if not warm_mask.any( ):
            warm_mask = ( temp_series == temp_series.max( ) ).fillna( False )

        if not cold_mask.any( ):
            cold_mask = ( temp_series == temp_series.min( ) ).fillna( False )

        warm_peak_month = int( group.loc[ temp_series.idxmax( ), 'month' ] )

    # if there are no valid temp values, we can't define warm and cold months,
    # ... so set everything to NaN or False

    else:
        temp_mean = np.nan
        warm_mask = pd.Series( False, index = group.index )
        cold_mask = pd.Series( False, index = group.index )
        warm_peak_month = np.nan

    warm_oxygen = group.loc[ warm_mask, 'oxygen_mmean' ].dropna( )
    cold_oxygen = group.loc[ cold_mask, 'oxygen_mmean' ].dropna( )

    # if there are any valid oxygen values, we can find the lowest/worst month?
    if oxygen_series.notna( ).any( ):
        oxygen_nadir_month = int( group.loc[ oxygen_series.idxmin( ), 'month' ] )

    else:
        oxygen_nadir_month = np.nan

    # if we have both a warm peak month and an oxygen crash, we can calculate the phase lag in months
    if pd.notna( warm_peak_month ) and pd.notna( oxygen_nadir_month ):
        temp_oxygen_phase_lag = float( ( oxygen_nadir_month - warm_peak_month ) % 12 )

    else:
        temp_oxygen_phase_lag = np.nan

    # and then we can return all the signature stats for this station as a series
    # which will become columns in the final frame
    return pd.Series( { 
        'warm_peak_month': warm_peak_month,
        'oxygen_nadir_month': oxygen_nadir_month,
        'temp_oxygen_phase_lag': temp_oxygen_phase_lag,
        'warm_len_months': int( warm_mask.sum( ) ),
        'cold_len_months': int( cold_mask.sum( ) ),
        'warm_intensity_temp': float( group.loc[ warm_mask, 'water_temp_mmean' ].sub( temp_mean ).sum( ) ) if pd.notna( temp_mean ) and warm_mask.any( ) else np.nan,
        'cold_intensity_temp': float( temp_mean * int( cold_mask.sum( ) ) - group.loc[ cold_mask, 'water_temp_mmean' ].sum( ) ) if cold_mask.any( ) else 0.0,
        'warm_oxygen_mean': float( warm_oxygen.mean( ) ) if len( warm_oxygen ) > 0 else np.nan,
        'warm_oxygen_min': float( warm_oxygen.min( ) ) if len( warm_oxygen ) > 0 else np.nan,
        'cold_oxygen_mean': float( cold_oxygen.mean( ) ) if len( cold_oxygen ) > 0 else np.nan,
    } )
```

```python
# keep stations with at least 3 valid years
eligible_stations = station_baseline_window.loc[ 
    station_baseline_window[ 'n_valid_years' ] >= 5,
    [ 'region', 'station' ]
]

# pull only rows from each station's selected baseline years
water_base = water_base.merge( 
    station_baseline_years[ [ 'region', 'station', 'year' ] ],
    on = [ 'region', 'station', 'year' ],
    how = 'inner'
)

# then keep only stations that passed the >=5-year minimum
water_base = water_base.merge( eligible_stations, on = [ 'region', 'station' ], how = 'inner' )
#eligible_stations.sample(5)
#eligible_stations.describe( ).T
#base.sample( 10 )
```

```python
# okay, now build station character features from the selected baseline window
water_base[ 'month' ] = water_base[ 'datetime' ].dt.month
# get mean of each year per station, then the mean of those five years per station
# yearly means first ( equal-year weighting )
annual = ( 
    water_base
    .groupby( [ 'region', 'station', 'year' ], as_index = False )
    .agg( 
        water_temp_ymean = ( 'water_temp', 'mean' ),
        salinity_ymean = ( 'salinity', 'mean' ),
        oxygen_ymean = ( 'oxygen', 'mean' ),
        saturation_ymean = ( 'oxy_saturation', 'mean' ),
        ph_ymean = ( 'ph', 'mean' ),
        depth_ymean = ( 'depth', 'mean' )
    )
)

# each station gets its own, yo
annual_means = ( 
    annual
    .groupby( [ 'region', 'station' ], as_index = False )
    .agg( 
        mean_annual_water_temp = ( 'water_temp_ymean', 'mean' ),
        mean_annual_salinity = ( 'salinity_ymean', 'mean' ),
        mean_annual_oxygen = ( 'oxygen_ymean', 'mean' ),
        mean_annual_saturation = ( 'saturation_ymean', 'mean' ),
        mean_annual_ph = ( 'ph_ymean', 'mean' ),
        mean_annual_depth = ( 'depth_ymean', 'mean' ),
        n_years_water_temp = ( 'water_temp_ymean', 'count' ),
        n_years_salinity = ( 'salinity_ymean', 'count' ),
        n_years_oxygen = ( 'oxygen_ymean', 'count' ),
        n_years_saturation = ( 'saturation_ymean', 'count' ),
        n_years_ph = ( 'ph_ymean', 'count' ),
        n_years_depth = ( 'depth_ymean', 'count' )
    )
)

# annual.sample( 8 ).T
# annual_means.sample( 8 ).T
```

```python
# seasonal amplitudes from monthly climatology
monthly = ( 
    water_base
    .groupby( [ 'region', 'station', 'month' ], as_index = False )
    .agg( 
        water_temp_mmean = ( 'water_temp', 'mean' ),
        salinity_mmean = ( 'salinity', 'mean' ),
        oxygen_mmean = ( 'oxygen', 'mean' ),
        saturation_mmean = ( 'oxy_saturation', 'mean' ),
        ph_mmean = ( 'ph', 'mean' ),
        depth_mmean = ( 'depth', 'mean' )
    )
)
#monthly.sample( 8 ).T
```

```python
# build both a simple amplitude table and a richer warm / cold season signature table
seasonal_amp = ( 
    monthly
    .groupby( [ 'region', 'station' ], as_index = False )
    # lambda (mini-function) to get max - min for each variable's monthly means, which is the seasonal amplitude
    .agg( 
        seasonal_amp_water_temp = ( 'water_temp_mmean', lambda s: s.max( ) - s.min( ) ),
        seasonal_amp_salinity = ( 'salinity_mmean', lambda s: s.max( ) - s.min( ) ),
        seasonal_amp_oxygen = ( 'oxygen_mmean', lambda s: s.max( ) - s.min( ) ),
        seasonal_amp_saturation = ( 'saturation_mmean', lambda s: s.max( ) - s.min( ) ),
        seasonal_amp_ph = ( 'ph_mmean', lambda s: s.max( ) - s.min( ) ),
        seasonal_amp_depth = ( 'depth_mmean', lambda s: s.max( ) - s.min( ) )
    )
)
#seasonal_amp.sample( 8 ).T
```

```python
# and then the richer seasonal signature table with warm/cold peaks and lags and all that tmbg
seasonal_signature = ( 
    monthly
    .groupby( [ 'region', 'station' ] )
    .apply( build_season_signature )  # <-- the big def: from above
    .reset_index( )
)
#seasonal_signature.sample( 8 ).T
```

```python
# depth summary ( useful if sensor environment differs by station )
# let's see if the recorders are variable depth per station, or predictable?
# could be useless
depth_stats = ( 
    water_base
    .groupby( [ 'region', 'station' ], as_index = False )
    .agg( 
        median_depth = ( 'depth', 'median' ), # meters, btw..
        p10_depth = ( 'depth', lambda s: s.quantile( 0.10 ) ),
        p90_depth = ( 'depth', lambda s: s.quantile( 0.90 ) )
    )
)
# iqr here basically means how much the depth varies at a station, 
# which is a proxy for whether it's in a tidal area or not
depth_stats[ 'iqr_depth' ] = depth_stats[ 'p90_depth' ] - depth_stats[ 'p10_depth' ]
#depth_stats.sample( 8 ).T
```

```python
# just so we know which stations have more or less data, 
# which could be a factor in how much we trust 'em
coverage = ( 
    water_base
    .groupby( [ 'region', 'station' ], as_index = False )
    .agg( 
        n_obs_total = ( 'datetime', 'size' ),
        n_years_present = ( 'year', 'nunique' )
    )
)
#coverage.sample( 8 ).T
```

```python
# lets snap these together into a baseline ...
station_baseline = ( 
    annual_means
    # btw, left joins so that we keep all rows in annual_means 'on the left', and just have blanks
    # 'on the right' whenever there is not a matching row in the below tables ...
    .merge( seasonal_amp, on = [ 'region', 'station' ], how = 'left' )
    .merge( seasonal_signature, on = [ 'region', 'station' ], how = 'left' )
    .merge( depth_stats, on = [ 'region', 'station' ], how = 'left' )
    .merge( coverage, on = [ 'region', 'station' ], how = 'left' )
    .merge( 
        station_baseline_window[ [ 
            'region',
            'station',
            'baseline_start_year',
            'baseline_end_year',
            'n_valid_years',
            'mean_year_coverage',
            'insufficient_baseline'
        ] ],
        on = [ 'region', 'station' ],
        how = 'left'
    )
    .sort_values( [ 'region', 'station' ] )
    .reset_index( drop = True )
)

station_baseline[ 'character_imputed' ] = False
#baseline.sample( 6 ).T
```

```python
# let's clean out temporary tables as they take up memory (and the data is big)
del( annual )
del( annual_means )
del( monthly )
del( seasonal_amp )
del( seasonal_signature )
del( depth_stats )
del( coverage )
del( station_baseline_window )
del( station_baseline_years )
```

### Stage 1.1 - Regime Clustering

```python
# now... for some clusters!
# selected features:
# - broad baseline hydroclimate structure only
# - keep the richer warm / cold season signature fields in baseline,
#   but do not let them define the primary regime geometry
# separate for re-use
regime_feature_cols = [ 
    'mean_annual_water_temp',
    'seasonal_amp_water_temp',
    'mean_annual_salinity',
    'seasonal_amp_salinity',
    'mean_annual_oxygen',
    'seasonal_amp_oxygen',
    'mean_annual_depth'
    # should we consider iqr_depth as a standin for 'tidal action' ???
]

# those plus the lookup indexes ...
# btw, copying to avoid potential in-memory damage to original table
regime = station_baseline[ [ 'region', 'station' ] + regime_feature_cols ].copy( )

# simple missing-value strategy: region median, then global median
# this is a bit of a band-aid, but the clustering is just for exploratory purposes
# and we don't want to lose stations with patchy data entirely
for col in regime_feature_cols:
    regime[ col ] = regime.groupby( 'region' )[ col ].transform( lambda s: s.fillna( s.median( ) ) )
    regime[ col ] = regime[ col ].fillna( regime[ col ].median( ) )

#regime.sample( 6 ).T
```

```python
# standard scalar for clustering, since the features are on very different scales 
# and we want them to have equal weight in distance calculations
# re: salinity vs temp vs oxygen ... all very different and could weight the PCA unfairly
scaler_regime = StandardScaler( )
scaled_regime = scaler_regime.fit_transform( regime[ regime_feature_cols ] )
```

```python
# k scan for elbow + silhouette
k_scan_rows = [ ]
k_min_interpretable = 3
k_max_interpretable = 6 
# if we did this again, might try 9 -> # saline(3) * temp(3) max (see plot)

# how lets go through a rational set of possible cluster counts
# and see what the inertia and silhouette score look like for each
for k in range( k_min_interpretable, k_max_interpretable + 1 ):
    # life, universe, and everything
    km_scan = KMeans( n_clusters = k, random_state = 42, n_init = 20 )
    labels_scan = km_scan.fit_predict( scaled_regime )

    k_scan_rows.append( { 
        'k': k,
        'inertia': float( km_scan.inertia_ ),
        'silhouette': float( silhouette_score( scaled_regime, labels_scan ) )
    } )

k_scan_regime = pd.DataFrame( k_scan_rows )
# calculate the drop in inertia from each additional cluster
k_scan_regime[ 'inertia_drop' ] = k_scan_regime[ 'inertia' ].shift( 1 ) - k_scan_regime[ 'inertia' ]
# and the percent drop relative to the previous inertia
k_scan_regime[ 'inertia_drop_pct' ] = k_scan_regime[ 'inertia_drop' ] / k_scan_regime[ 'inertia' ].shift( 1 )

# and now ... the moment of truth: which k is best?
recommended_k = int( k_scan_regime.loc[ k_scan_regime[ 'silhouette' ].idxmax( ), 'k' ] )
#print( f"recommended k: { recommended_k }" )
del( k_min_interpretable )
```

### Stage 2 - Cluster Diagnostics

These checks justify the chosen cluster count before the notebook moves into daily features.

### Inertia
- Sum of squared distances from each point to its assigned cluster center
- Lower is always better, but it always drops as K increases
- So you do not pick the minimum directly; you look for the “elbow,” where adding more clusters stops buying much improvement
- It is mainly a compactness / diminishing-returns measure

### Silhouette
- Measures how well-separated clusters are, roughly comparing within-cluster tightness to nearest-other-cluster distance
- Ranges from -1 to 1
- Higher is better
- Near 1: points are well matched to their own cluster and far from others
- Near 0: clusters overlap
- Negative: many points may be assigned to the wrong cluster
- This is more directly useful for picking K

```python

# in case we want a look at the elbow and silhouette charts, 
# which are the basis for our k choice
fig, axes = plot.subplots( 1, 2, figsize = ( 12, 3 ) )

axes[ 0 ].plot( 
    k_scan_regime[ 'k' ],
    k_scan_regime[ 'inertia' ],
    marker = 'o'
)
axes[ 0 ].set_title( 'Inertia (K)' )

axes[ 1 ].plot( 
    k_scan_regime[ 'k' ],
    k_scan_regime[ 'silhouette' ],
    marker = 'o'
)
axes[ 1 ].set_title( 'Silhouette (K)' )

plot.tight_layout( )
plot.show( ) 

# cleanup
del( k_scan_rows, k_scan_regime )
```

```python
# and then we can use those z-scores to bucket each cluster's profile 
# on each feature into 'low', 'mid', and 'high' buckets


def bucket_tag( value, low_label, mid_label, high_label, threshold = 0.4 ):
    if pd.isna( value ):
        return mid_label

    if value >= threshold:
        return high_label

    if value <= -threshold:
        return low_label

    return mid_label
```

```python
# and then we can combine those bucket tags into readable cluster names and notes


def build_cluster_name( z_row ):
    temp_tag = bucket_tag( z_row[ 'mean_annual_water_temp' ], 'Cool', 'Temp', 'Warm' )
    sal_tag = bucket_tag( z_row[ 'mean_annual_salinity' ], 'Fresh', 'Brackish', 'Saline' )
    oxy_tag = bucket_tag( z_row[ 'mean_annual_oxygen' ], 'Low O2', 'Mid O2', 'High O2' )

    amp_tag = bucket_tag( z_row[ 'seasonal_amp_water_temp' ], 'Stable', 'Mixed', 'Seasonal' )
    depth_tag = bucket_tag( z_row[ 'mean_annual_depth' ], 'Shallow', 'Mid', 'Deep' )

    name = f'{ temp_tag } / { sal_tag } / { oxy_tag }'
    short_name = f'{ temp_tag }-{ sal_tag }'
    note = f'{ amp_tag }, { depth_tag }'

    return pd.Series( { 
        'cluster_name': name, 
        'cluster_label': short_name, 
        'cluster_note': note,
        #'cluster_temp': z_row[ 'mean_annual_water_temp' ],
        #'cluster_salinity': z_row[ 'mean_annual_salinity' ],
        #'cluster_oxygen': z_row[ 'mean_annual_oxygen' ]
    } )
```

# Phase 4: Multivariate Analysis

Before trusting the regime labels, we look at how the baseline features relate to each other. Strong correlations are expected in estuary data, but the clustering should still separate interpretable combinations of temperature, salinity, oxygen, and depth.

```python
# pretty correlation matrix + paired scatterplots

#   - the DIAGONAL (top-left to bottom-right) shows the variable name for that row/column
#   - the UPPER-RIGHT triangle shows a scatterplot of the two variables at that intersection
#   - the LOWER-LEFT triangle shows the pearson r correlation value as a colored tile

# the tile color comes from a coolwarm colormap anchored at -1 (blue) through 0 (white)
# to +1 (red), so you can read the strength and direction of a relationship at a glance
# before you even read the number.  a colorbar on the right side gives the key.

# this is an early exploratory look — we just want to see which variables move together
# before we start building models or digging into regimes and baselines.

# major original metrics
correlation_cols = [
    'air_temp',
    'water_temp',
    'precipitation',
    'wind_speed',
    'salinity',
    'oxygen',
    'ph',
]

# readable two-line labels for each
correlation_label_map = {
    'air_temp':      'Air Temp\n( C )',
    'water_temp':    'Water Temp\n( C )',
    'precipitation': 'Precip\n( mm / hr )',
    'wind_speed':    'Wind\n( m / s )',
    'salinity':      'Salinity\n( PSU )',
    'oxygen':        'Oxygen\n( mg / L )',
    'ph':            'pH',
}

# make sure everything is numeric
correlation_df = water[ correlation_cols ].apply( pd.to_numeric, errors = 'coerce' )
correlation_display = correlation_df.rename( columns = correlation_label_map ) # rename stuff

# just make sure we have only complete rows
pairplot_df = correlation_df.dropna( )

# reduce points just not to waste clock cycles on things we can't see
pairplot_n_max = 4000
if len( pairplot_df ) > pairplot_n_max:
    pairplot_df = pairplot_df.sample( pairplot_n_max, random_state = 42 )

# apply the friendly labels to the sampled frame — this is what the grid will see
pairplot_display = pairplot_df.rename( columns = correlation_label_map )

# prep colors and range
pairplot_cmap = plot.get_cmap( 'coolwarm' )
pairplot_norm = plot.Normalize( -1, 1 )

# seaborn's PairGrid lets us hand it a custom function to paint each lower-triangle
# the function:
#   1. computes pearson r for the valid (non-NaN) observations in that pair
#   2. picks a background color from the coolwarm map based on that r value
#   3. writes the r value as large bold text centered in the tile
#   4. strips out all ticks and border lines so the color and number are all you see
def draw_corr_tile( x, y, **kws ):
    """draw a single correlation tile for the lower triangle of the pair grid."""

    ax = plot.gca( )

    # wrap in pd.Series so .notna() works reliably regardless of what seaborn passes in
    x = pd.Series( x )
    y = pd.Series( y )

    # we can only compute a correlation if both columns have at least two valid rows together
    valid = x.notna( ) & y.notna( )

    if valid.sum( ) < 2:
        # not enough data for this pair — mark it as missing
        corr_value = np.nan

    else:
        # np.corrcoef returns a 2x2 matrix; [0,1] is the off-diagonal r value we want
        corr_value = float( np.corrcoef( x[ valid ], y[ valid ] )[ 0, 1 ] )

    # choose tile background: light gray for missing data, colormap color for a real r
    tile_color = '#f4f4f4' if pd.isna( corr_value ) else pairplot_cmap( pairplot_norm( corr_value ) )
    ax.set_facecolor( tile_color )

    # write the correlation number (or "NA") centered in the tile in large bold text
    ax.text(
        0.5,
        0.5,
        'NA' if pd.isna( corr_value ) else f'{ corr_value:.2f}',
        ha          = 'center',
        va          = 'center',
        fontsize    = 30,
        fontweight  = 'bold',
        transform   = ax.transAxes,
        color       = 'black',
    )

    # remove ticks and spines — we want a clean colored square with just a number
    ax.set_xticks( [ ] )
    ax.set_yticks( [ ] )

    for spine in ax.spines.values( ):
        spine.set_visible( False )

# sns.PairGrid creates an N×N grid of subplots (one per variable pair)
#   map_upper and map_lower each take a drawing function and apply it to
#   every cell in the upper-right and lower-left triangles respectively
#   upper-right triangle → seaborn scatter plots (raw data relationship)
#   lower-left triangle  → our custom draw_corr_tile (colored r value)
#   diagonal             → handled separately below (variable name label)
phase1_pair_grid = sns.PairGrid( pairplot_display, height = 1.9, aspect = 1 )

# upper-right: small, semi-transparent dots so dense clouds don't fully obscure structure
phase1_pair_grid.map_upper(
    sns.scatterplot,
    s         = 12,
    alpha     = 0.25,
    linewidth = 0,
)

# lower-left: our colored correlation tiles
phase1_pair_grid.map_lower( draw_corr_tile )

# seaborn normally puts a histogram or KDE on the diagonal
# we skip that and instead write the variable name centered in each diagonal cell so the grid is self-labeling
for row_idx, col_name in enumerate( pairplot_display.columns ):

    # grab the subplot that sits on the diagonal for this variable
    diag_ax = phase1_pair_grid.axes[ row_idx, row_idx ]

    # very light background so the label cell reads as distinct from the data cells
    diag_ax.set_facecolor( '#fbfbfb' )

    # write the variable name centered in the cell
    diag_ax.text(
        0.5,
        0.5,
        col_name,
        ha        = 'center',
        va        = 'center',
        fontsize  = 20,
        wrap      = True,
        transform = diag_ax.transAxes,
    )

    # no ticks or borders — keep the diagonal labels as clean as the correlation tiles
    diag_ax.set_xticks( [ ] )
    diag_ax.set_yticks( [ ] )

    for spine in diag_ax.spines.values( ):
        spine.set_visible( False )

# seaborn adds axis labels by default - we suppress the y-axis label text because
# the diagonal already tells you which variable is which
for ax in phase1_pair_grid.axes.flat:
    if ax is None:
        continue
    ax.tick_params( labelsize = 9 )
    ax.xaxis.label.set_size( 12 )
    ax.yaxis.label.set_size( 12 )
    ax.set_ylabel( '' )

# wspace and hspace near zero makes the cells butt up against each other so the
# grid reads as one unified figure rather than a collection of separate plots
# the right margin is opened up to make room for the colorbar.
phase1_pair_grid.fig.subplots_adjust( top = 0.93, right = 0.9, wspace = 0.04, hspace = 0.04 )
phase1_pair_grid.fig.suptitle(
    'Air + Water Variable Matrix | Scatter Upper, Correlation Lower',
    y = 0.995
)

# the colorbar lives in a thin strip on the right edge of the figure.
# add_axes takes [left, bottom, width, height] in figure-fraction coordinates.
phase1_colorbar_ax = phase1_pair_grid.fig.add_axes( [ 0.92, 0.2, 0.02, 0.6 ] )

# ScalarMappable is what matplotlib needs to attach a colormap to a colorbar
# without a "real" plot backing it — we're manually creating the mapping here
# because the correlation tiles were drawn by our custom function, not by any
# standard matplotlib artist that would normally own a colorbar.
phase1_colorbar_sm = plot.cm.ScalarMappable( cmap = pairplot_cmap, norm = pairplot_norm )
phase1_colorbar_sm.set_array( [ ] )

phase1_pair_grid.fig.colorbar( phase1_colorbar_sm, cax = phase1_colorbar_ax, label = 'Pearson r' )

plot.show( )
```

```python
# and now, let's roll the main driver table down to daily rows
daily_baseline = water.dropna( subset = [ "datetime" ] )
daily_baseline [ "date" ] = daily_baseline[ "datetime" ].dt.floor( "D" )

daily_air = ( 
    daily_baseline 
    .groupby( [ "region", "station", "date" ], as_index = False )
    .agg( 
        air_temp = ( "air_temp", "mean" ),
        wind_speed = ( "wind_speed", "mean" ),
        solar = ( "solar_radiation", "mean" ),
        precip = ( "precipitation", "sum" ),
    )
    .sort_values( [ "region", "station", "date" ] )
)
#daily_air.describe( ).round( 3 )
```

### 5.1 Recurring Seasonal Gaps

These stations can look like regime transitions when the same missing season repeats across years. So we first summarize whether each station drops out for the same multi-month window again and again.


```python
# settings for this section
recurring_break_min_full_years = 3
recurring_break_min_missing_run_months = 2 # consecutive months missing, that is
recurring_break_year_share_threshold = 0.50
recurring_break_same_time_share_threshold = 0.60
```

```python
# and now, let's attach the cluster labels to the daily water table 
# as well, so we can use them in stage 9 for stratification and analysis


def first_full_calendar_year( ts ):
    if pd.isna( ts ):
        return np.nan

    if ts.month == 1 and ts.day == 1:
        return int( ts.year )

    return int( ts.year ) + 1

# this is basically the opposite of the above func, but we need both 
# to define the full calendar year window for each station's baseline period


def last_full_calendar_year( ts ):
    if pd.isna( ts ):
        return np.nan

    if ts.month == 12 and ts.day == 31:
        return int( ts.year )

    return int( ts.year ) - 1
```

```python
# we need to determine if a station is missing entire months worth of data


def summarize_missing_month_run( month_obs_values ):
    max_run = 0
    best_start_month = np.nan
    current_run = 0
    current_start = None

    # loop through the monthly observation counts for a station, 
    # and track runs of zero-observation months
    for month_idx, obs_days in enumerate( month_obs_values, start = 1 ):
        if obs_days == 0:
            if current_run == 0:
                current_start = month_idx
            current_run += 1
            if current_run > max_run:
                max_run = current_run
                best_start_month = current_start

        else:
            current_run = 0
            current_start = None

    # if the longest run of missing months is shorter than our threshold 
    # for a 'recurring break', then we can ignore it and set the start month to NaN
    if max_run < recurring_break_min_missing_run_months:
        best_start_month = np.nan

    return pd.Series( 
        { 
            "max_consecutive_missing_months": int( max_run ),
            "long_gap_start_month": best_start_month,
            "has_long_gap": bool( max_run >= recurring_break_min_missing_run_months )
        }
    )
```

```python
# and then we can use those to define the full calendar year window for each station's baseline period


def summarize_station_year_gap( frame ):
    frame = frame.sort_values( "month" )
    run_summary = summarize_missing_month_run( frame[ "n_obs_days" ].tolist( ) )
    return pd.Series( 
        { 
            "n_missing_whole_months": int( frame[ "missing_whole_month" ].sum( ) ),
            "n_obs_months": int( ( ~frame[ "missing_whole_month" ] ).sum( ) ),
            "max_consecutive_missing_months": int( run_summary[ "max_consecutive_missing_months" ] ),
            "long_gap_start_month": run_summary[ "long_gap_start_month" ],
            "has_long_gap": bool( run_summary[ "has_long_gap" ] )
        }
    )
```

```python
# for this part, we need to know how many days each station is observed in each month of each year, 
# so we can identify any recurring gaps that would break the assumption of a continuous baseline period
daily_presence = daily_air[ [ "region", "station", "date" ] ].drop_duplicates( )
daily_presence[ "year" ] = daily_presence[ "date" ].dt.year
daily_presence[ "month" ] = daily_presence[ "date" ].dt.month
```

```python
# lets get the span of years observed at each station, 
# and the number of unique years observed, 
# so we can identify stations with enough coverage to even consider for a baseline
station_span_p2 = ( 
    daily_presence
    .groupby( [ "region", "station" ], as_index = False )
    .agg( 
        first_obs_date = ( "date", "min" ),
        last_obs_date = ( "date", "max" ),
        n_years_obs = ( "year", "nunique" )
    )
)
station_span_p2[ "first_full_year" ] = station_span_p2[ "first_obs_date" ].apply( first_full_calendar_year )
station_span_p2[ "last_full_year" ] = station_span_p2[ "last_obs_date" ].apply( last_full_calendar_year )
station_span_p2[ "n_full_years" ] = np.maximum( 
    0,
    # add 1 because if the first and last full years are the same, 
    # that still counts as one full year of coverage
    station_span_p2[ "last_full_year" ] - station_span_p2[ "first_full_year" ] + 1,
).astype( int )
#station_span_p2.describe( )
#station_span_p2.sample( 5 ).T
```

```python
full_year_rows = [ ]
# now we can loop through each station, and if it has at least one full calendar year of data, 
# we can build rows for each month of each full calendar year in that station's observed span
for row in station_span_p2.itertuples( index = False ):
    if int( row.n_full_years ) <= 0:
        continue

    for year in range( int( row.first_full_year ), int( row.last_full_year ) + 1 ):
        full_year_rows.append( 
            { 
                "region": row.region,
                "station": row.station,
                "year": int( year )
            }
        )

# and then we can use that to build a full station-year-month grid for any station with at least one full calendar year
if full_year_rows:
    station_full_years_p2 = pd.DataFrame( full_year_rows )
    month_grid = pd.DataFrame( { "month": np.arange( 1, 13, dtype = int ) } )
    station_full_years_p2[ "_tmp" ] = 1
    month_grid[ "_tmp" ] = 1
    station_year_month_grid_p2 = station_full_years_p2.merge( month_grid, on = "_tmp", how = "inner" ).drop( columns = [ "_tmp" ] )

else:
    station_year_month_grid_p2 = pd.DataFrame( columns = [ "region", "station", "year", "month" ] )

#station_year_month_grid_p2.sample( 6 ).T
```

```python
# and now we can count the number of days observed in each month of each year for each station,
monthly_obs_counts_p2 = ( 
    daily_presence
    .groupby( [ "region", "station", "year", "month" ], as_index = False )
    .agg( n_obs_days = ( "date", "nunique" ) )
)

# and merge that with the full station-year-month grid, 
# so that we have a row for every month of every year of every station's observed span
station_year_month_p2 = station_year_month_grid_p2.merge( 
    monthly_obs_counts_p2,
    on = [ "region", "station", "year", "month" ],
    how = "left"
)
station_year_month_p2[ "n_obs_days" ] = station_year_month_p2[ "n_obs_days" ].fillna( 0 ).astype( int )
station_year_month_p2[ "missing_whole_month" ] = station_year_month_p2[ "n_obs_days" ] == 0

#station_year_month_grid_p2.sample( 5 ).T
```

```python
# now, we can summarize the gap structure of each station-year based on the monthly obs counts,
station_year_gap_summary_p2 = ( 
    station_year_month_p2
    .groupby( [ "region", "station", "year" ] )
    .apply( summarize_station_year_gap )
    .reset_index( )
)
#station_year_gap_summary_p2.sample( 5 ).T
#station_year_gap_summary_p2.describe( ).T
```

```python
long_gap_years_p2 = station_year_gap_summary_p2.loc[ station_year_gap_summary_p2[ "has_long_gap" ] ]

# so lets find the most common month for a long gap to start, for each station
long_gap_mode_p2 = ( 
    long_gap_years_p2
    .groupby( [ "region", "station", "long_gap_start_month" ], as_index = False )
    .agg( n_years_same_start = ( "year", "nunique" ) )
    .sort_values( [ "region", "station", "n_years_same_start", "long_gap_start_month" ], ascending = [ True, True, False, True ] )
    .drop_duplicates( subset = [ "region", "station" ] )
    .rename( columns = { "long_gap_start_month": "modal_long_gap_start_month" } )
)
#long_gap_mode_p2.describe( ).T
#long_gap_mode_p2.sample( 6 ).T
```

```python
long_gap_mode_p2.sample( 6 ).T
```

```python
# and then we can merge all those gap summary stats together to get a 
# station-level picture of the gap structure of each station's observed record
# which will decisions about how to handle gaps in the baseline period for each station
seasonal_break_summary = station_span_p2.merge( 
    ( 
        station_year_gap_summary_p2
        .groupby( [ "region", "station" ], as_index = False )
        .agg( 
            median_missing_whole_months = ( "n_missing_whole_months", "median" ),
            median_max_consecutive_missing_months = ( "max_consecutive_missing_months", "median" ),
            n_full_years_long_gap = ( "has_long_gap", "sum" ),
            pct_full_years_with_long_gap = ( "has_long_gap", "mean" )
        )
    ),
    on = [ "region", "station" ],
    how = "left",
)
seasonal_break_summary = seasonal_break_summary.merge( long_gap_mode_p2, on = [ "region", "station" ], how = "left" )

# fill in some gaps in the gap summary stats with zeros, 
# since if there are no long gaps, for example, then the median missing months 
# and max consecutive missing months should be zero, not null
for col in [ 
    "median_missing_whole_months",
    "median_max_consecutive_missing_months",
    "n_full_years_long_gap",
    "pct_full_years_with_long_gap",
    "n_years_same_start",
]:
    seasonal_break_summary[ col ] = seasonal_break_summary[ col ].fillna( 0 )
    
#seasonal_break_summary.describe( ).T
```

```python
# try to establish the MODE of a long gap
# if it becomes repetative, it's clear a station has 'seasons off' from their work
seasonal_break_summary[ "modal_long_gap_start_month" ] = seasonal_break_summary[ "modal_long_gap_start_month" ].fillna( 0 ).astype( int )
seasonal_break_summary[ "modal_long_gap_start_share" ] = np.where( 
    seasonal_break_summary[ "n_full_years_long_gap" ] > 0,
    seasonal_break_summary[ "n_years_same_start" ] / seasonal_break_summary[ "n_full_years_long_gap" ],
    0.0,
)

# how many breaks is too many breaks? 
seasonal_break_summary[ "seasonal_break_station" ] = ( 
    ( seasonal_break_summary[ "n_full_years" ] >= recurring_break_min_full_years )
    & ( seasonal_break_summary[ "pct_full_years_with_long_gap" ] >= recurring_break_year_share_threshold )
    & ( seasonal_break_summary[ "modal_long_gap_start_share" ] >= recurring_break_same_time_share_threshold )
)

# and finally, we can pull out the keys of the stations with seasonal breaks for later use in stage 9,
stations_taking_seasonal_breaks = ( 
    seasonal_break_summary
    .loc[ seasonal_break_summary[ "seasonal_break_station" ], [ "region", "station" ] ]
    .drop_duplicates( )
    .sort_values( [ "region", "station" ] )
    .reset_index( drop = True )
)

# cleanup
del( 
    station_span_p2, 
    station_full_years_p2, 
    month_grid, 
    station_year_month_grid_p2, 
    monthly_obs_counts_p2, 
    station_year_month_p2, 
    station_year_gap_summary_p2, 
    long_gap_years_p2, 
    long_gap_mode_p2, 
    seasonal_break_summary
)
```

### 5.3 Five-Year Rolling Monthly Regime Table

We classify on monthly 5-year rolling summaries instead of raw daily rows. That is more tolerant of seasonal sampling gaps and gives a smoother regime path.


```python
# lets set aside a copy to 'water down' to a 5 year rolling mean
fiveyear_water = daily_water.copy( )
fiveyear_water[ "date" ] = pd.to_datetime( fiveyear_water[ "date" ], errors = "coerce" )
fiveyear_water = fiveyear_water.dropna( subset = [ "date" ] )
fiveyear_water = fiveyear_water.sort_values( [ "region", "station", "date" ] )
fiveyear_water[ "month" ] = fiveyear_water[ "date" ].dt.to_period( "M" ).dt.to_timestamp( )
#fiveyear_water.sample( 4 ).T
```

```python
# metrics to track (repeated)
fiveyear_water_metrics = [ "water_temp", "salinity", "oxygen", "ph", "depth" ]
```

```python
# lets get a monthly mean sense of each station...
monthly_water = ( 
    fiveyear_water
    .groupby( [ "region", "station", "month" ], as_index = False )[ fiveyear_water_metrics ]
    .mean( )
    .sort_values( [ "region", "station", "month" ] )
)
#monthly_water.describe( )
```

```python
# roll those monthly means up to a 5 year rolling mean, 
# which is the main form of smoothing we'll use for the baselines 
for metric in fiveyear_water_metrics:
    monthly_water[ f"{ metric }_roll5y" ] = ( 
        monthly_water
        .groupby( [ "region", "station" ] )[ metric ]
        .transform( lambda values: values.rolling( window = 60, min_periods = 36 ).mean( ) )
    )

# did it get anything?
monthly_water[ "has_any_data" ] = monthly_water[ fiveyear_water_metrics ].notna( ).any( axis = 1 ).astype( int )
monthly_water[ "months_with_any_data_5y" ] = ( 
    monthly_water
    .groupby( [ "region", "station" ] )[ "has_any_data" ]
    .transform( lambda values: values.rolling( window = 60, min_periods = 1 ).sum( ) )
)
monthly_water[ "coverage_ratio_5y" ] = monthly_water[ "months_with_any_data_5y" ] / 60.0
```

```python
# and then we can merge those 5-year rolling means and coverage metrics back into the station-year-month grid,
roll5y_cols = [ f"{ metric }_roll5y" for metric in fiveyear_water_metrics ]
monthly_roll_cols = [ "region", "station", "month" ] + roll5y_cols + [ "months_with_any_data_5y", "coverage_ratio_5y" ]
fiveyear_water = fiveyear_water.merge( 
    monthly_water[ monthly_roll_cols ], 
    on = [ "region", "station", "month" ], 
    how = "left"
)
```

```python
# the monthly estimate of regime can be 
feature_metrics = [ 
    "mean_annual_water_temp",
    "mean_annual_salinity",
    "mean_annual_oxygen",
    "mean_annual_depth",
]

ref = station_baseline[ [ "cluster" ] + feature_metrics ].dropna( ).copy( )

# make sure stuff with large values like temp or saturation
# dont dominate things with low values like depth or pH
scaler_p5 = StandardScaler( ).fit( ref[ feature_metrics ] )

new_regime = pd.DataFrame( 
    scaler_p5.transform( ref[ feature_metrics ] ),
    columns = feature_metrics,
    index = ref.index,
)
new_regime[ "cluster" ] = ref[ "cluster" ].values

new_centroids = ( 
    new_regime
    .groupby( "cluster", as_index = True )[ feature_metrics ]
    .mean( )
    .sort_index( )
)

coverage_threshold = 0.70
min_features_required = 3
feature_mean = pd.Series( scaler_p5.mean_, index = feature_metrics )
feature_scale = pd.Series( scaler_p5.scale_, index = feature_metrics )
```

### 5.4 Nearest-Centroid Classification

Each monthly 5-year profile is compared against the baseline regime centroids from stage 1. We only classify rows with enough rolling coverage and at least three usable features.


```python
#  now we can write a function to classify each monthly row based 
# on the distance of its available features to the centroids


def classify_month_row( row ):

    # if the station-month doesn't have enough coverage in the past 5 years
    # then we can't really trust the rolling mean
    # so we should just mark this as unclassified due to insufficient coverage
    coverage = row.get( "coverage_ratio_5y", np.nan )
    if pd.isna( coverage ) or coverage < coverage_threshold:
        return pd.Series( 
            { 
                "implied_cluster": np.nan,
                "dist_best": np.nan,
                "dist_second": np.nan,
                "margin": np.nan,
                "n_features_used": 0,
                "assignment_state": "insufficient_coverage",
            }
        )

    # then we can check which features are actually available for this station-month
    available = [ col for col in feature_metrics if pd.notna( row.get( col, np.nan ) ) ]
    if len( available ) < min_features_required:
        return pd.Series( 
            { 
                "implied_cluster": np.nan,
                "dist_best": np.nan,
                "dist_second": np.nan,
                "margin": np.nan,
                "n_features_used": len( available ),
                "assignment_state": "insufficient_features",
            }
        )

    # if we have enough coverage and features, then we can calculate the z-scores
    values = pd.to_numeric( row[ available ], errors = "coerce" )
    scales = pd.to_numeric( feature_scale[ available ], errors = "coerce" ).replace( 0, np.nan )
    means = pd.to_numeric( feature_mean[ available ], errors = "coerce" )
    z_values = ( values - means ) / scales

    # if any of the z-scores are NaN, then we have a scaling issue and can't classify
    if z_values.isna( ).any( ):
        return pd.Series( 
            { 
                "implied_cluster": np.nan,
                "dist_best": np.nan,
                "dist_second": np.nan,
                "margin": np.nan,
                "n_features_used": len( available ),
                "assignment_state": "scaling_issue",
            }
        )

    # and then we can calculate the distance from this station-month to each of the centroids
    centroid_slice = new_centroids[ available ]
    centroid_array = np.asarray( centroid_slice.to_numpy( ), dtype = float )
    z_array = np.asarray( z_values.to_numpy( ), dtype = float )
    distances = np.sqrt( np.sum( ( centroid_array - z_array[ None, : ] ) ** 2, axis = 1 ) )

    # and find the closest centroid, and the distance to that centroid
    # and the margin to the second closest centroid
    best_idx = int( np.argmin( distances ) )
    best_cluster = int( centroid_slice.index[ best_idx ] )
    best_dist = float( distances[ best_idx ] )

    # get the second closest distance
    second_dist = float( np.partition( distances, 1 )[ 1 ] )
    margin = second_dist - best_dist

    # if we got this far, then we have a valid classification
    state = "classified_full" if len( available ) == len( feature_metrics ) else "classified_partial"

    return pd.Series( 
        { 
            "implied_cluster": best_cluster,
            "dist_best": best_dist,
            "dist_second": second_dist,
            "margin": margin,
            "n_features_used": len( available ),
            "assignment_state": state,
        }
    )
```

### 5.5 First Confirmed Transition Events

A station only counts as transitioning when the monthly implied regime persists for at least six months and clears a small distance margin over the next-best centroid.


```python
# sort!
p5_monthly = p5_monthly.sort_values( [ "region", "station", "date" ] ).copy( )

# flag any station-months where the implied cluster is different than the assigned cluster
p5_monthly[ "candidate_flip" ] = ( 
    p5_monthly[ "implied_cluster" ].notna( )
    & p5_monthly[ "cluster" ].notna( )
    & ( p5_monthly[ "implied_cluster" ] != p5_monthly[ "cluster" ] )
)

# look for runs of consecutive months where the implied cluster is different than the assigned cluster
monthly_run_id = ( 
    p5_monthly
    .groupby( [ "region", "station" ] )[ "implied_cluster" ]
    .transform( lambda values: values.ne( values.shift( ) ).cumsum( ) )
)

# how long are those runs of potential flips?
p5_monthly[ "run_len_months" ] = ( 
    p5_monthly
    .groupby( [ "region", "station", monthly_run_id ] )[ "implied_cluster" ]
    .transform( "size" )
)

# flag any station-months where the implied cluster is different than the assigned cluster
p5_monthly[ "flip_confirmed_monthly" ] = ( 
    p5_monthly[ "candidate_flip" ]
    & ( p5_monthly[ "run_len_months" ] >= 6 )
    & ( p5_monthly[ "margin" ] > 0.10 )
)
```

### 6.0 Feature Engineering Helper

This helper adds the interaction and delta features that stage 6 and stage 9 both reuse.


```python
def add_air_water_features( frame ):
    frame[ "date" ] = pd.to_datetime( frame[ "date" ], errors = "coerce" )
    frame[ "doy" ] = frame[ "date" ].dt.dayofyear
    frame[ "month" ] = frame[ "date" ].dt.month

    if { "air_temp", "water_temp_baseline" }.issubset( frame.columns ):
        frame[ "air_temp_minus_water_temp_baseline" ] = frame[ "air_temp" ] - frame[ "water_temp_baseline" ]

    if { "air_temp_r7d", "water_temp_baseline" }.issubset( frame.columns ):
        frame[ "air_temp_r7d_minus_water_temp_baseline" ] = frame[ "air_temp_r7d" ] - frame[ "water_temp_baseline" ]

    if { "air_temp_r28d", "water_temp_baseline" }.issubset( frame.columns ):
        frame[ "air_temp_r28d_minus_water_temp_baseline" ] = frame[ "air_temp_r28d" ] - frame[ "water_temp_baseline" ]

    if { "air_temp_r1d", "air_temp_r28d" }.issubset( frame.columns ):
        frame[ "air_temp_r1d_minus_air_temp_r28d" ] = frame[ "air_temp_r1d" ] - frame[ "air_temp_r28d" ]

    if { "air_temp_r7d", "air_temp_r28d" }.issubset( frame.columns ):
        frame[ "air_temp_r7d_minus_air_temp_r28d" ] = frame[ "air_temp_r7d" ] - frame[ "air_temp_r28d" ]

    if { "wind_speed", "air_temp" }.issubset( frame.columns ):
        frame[ "wind_speed_x_air_temp" ] = frame[ "wind_speed" ] * frame[ "air_temp" ]

    if { "solar", "air_temp" }.issubset( frame.columns ):
        frame[ "solar_x_air_temp" ] = frame[ "solar" ] * frame[ "air_temp" ]

    return frame
```

```python
# build out a baseline climate temperature by day of year
# we'll compare against this baseline, and a persistence baseline, 
# to see if we can beat them with the air features and/or the station context features
climate_by_day = ( 
    air_water_train_model
    .groupby( "doy", as_index = False )[ target_col ]
    .median( )
    .rename( columns = { target_col: "pred_climatology" } )
)

# attach that climatology to the holdout set, so we can compare against it
air_water_holdout_model[ "pred_climatology" ] = air_water_holdout_model[ "doy" ].map( climate_by_day.set_index( "doy" )[ "pred_climatology" ] )
global_target_median = float( air_water_train_model[ target_col ].median( ) )
air_water_holdout_model[ "pred_persistence" ] = air_water_holdout_model[ "pred_persistence" ].fillna( air_water_holdout_model[ "pred_climatology" ] ).fillna( global_target_median )
air_water_holdout_model[ "pred_climatology" ] = air_water_holdout_model[ "pred_climatology" ].fillna( global_target_median )

# and then we can set up some storage for the scores and models
y_train = air_water_train_model[ target_col ]
air_water_score_rows = [ ]
air_water_model_store = { }
air_water_feature_store = { }
air_water_fill_store = { }
air_water_pred_col_store = { 
    "naive_persistence": "pred_persistence",
    "naive_climatology_doy": "pred_climatology"
}

climate_by_day.sample( 5 ).T
```

### 6.2 Fit Candidate Models

This cell fits the ridge and gradient-boosting candidates and compares their raw holdout performance.

**note:** prediction error + alpha * sum( coefficient^2 ) ... akin to linear, but with guardrails

**from reading:**
"*ridge is a sensible model when you have climate and station features that overlap in information*", like:

- air temperature means and standard deviations
- wind, precip, solar summaries
- baseline station traits
- clustered or seasonal descriptors

e.g., "*find the best weights, but don’t trust huge weights unless the data really forces them*"

Not as good at nonlinear behavior and keeps features in the model that might have been selectably out?


```python
X_train_air = air_water_train_model[ air_features ]
air_fill_values = X_train_air.median( numeric_only = True )
X_train_air = X_train_air.fillna( air_fill_values )

# air-only model as a baseline, since air features are the most common and have the most coverage 
# - let's see how far we can get with just those before even adding context features
ridge_air_model = Ridge( alpha = 1.0 )
ridge_air_model.fit( X_train_air, y_train )
air_water_model_store[ "ridge_air_only" ] = ridge_air_model
air_water_feature_store[ "ridge_air_only" ] = air_features
air_water_fill_store[ "ridge_air_only" ] = air_fill_values
air_water_pred_col_store[ "ridge_air_only" ] = "pred_ridge_air_only"

# we can only build joint models on some rows that have at least some coverage, but let's see if that buys us anything
X_train_joint = air_water_train_model[ joint_features ] if joint_features else pd.DataFrame( index = air_water_train_model.index )
joint_fill_values = X_train_joint.median( numeric_only = True ) if joint_features else pd.Series( dtype = "float64" )
X_train_joint = X_train_joint.fillna( joint_fill_values ) if joint_features else X_train_joint

# note that we don't have enough data to really tune these models, 
# so we'll just pick some reasonable defaults and see how they do 
# - no hyperparameter tuning at this stage
if joint_features:
    ridge_air_context_model = Ridge( alpha = 1.0 )
    ridge_air_context_model.fit( X_train_joint, y_train )
    air_water_model_store[ "ridge_air_context" ] = ridge_air_context_model
    air_water_feature_store[ "ridge_air_context" ] = joint_features
    air_water_fill_store[ "ridge_air_context" ] = joint_fill_values
    air_water_pred_col_store[ "ridge_air_context" ] = "pred_ridge_air_context"

    hgb_air_context_model = HistGradientBoostingRegressor( 
        learning_rate = 0.05,
        max_depth = 6,
        max_iter = 250,
        random_state = 42, # everyone's favorite number
    )
    hgb_air_context_model.fit( X_train_joint, y_train )
    air_water_model_store[ "hgb_air_context" ] = hgb_air_context_model
    air_water_feature_store[ "hgb_air_context" ] = joint_features
    air_water_fill_store[ "hgb_air_context" ] = joint_fill_values
    air_water_pred_col_store[ "hgb_air_context" ] = "pred_hgb_air_context"

# and then we can see how those models do on the holdout set, compared to the baselines
y_holdout = air_water_holdout_model[ target_col ]
append_air_water_score( air_water_score_rows, "naive_persistence", y_holdout, air_water_holdout_model[ "pred_persistence" ] )
append_air_water_score( air_water_score_rows, "naive_climatology_doy", y_holdout, air_water_holdout_model[ "pred_climatology" ] )

# and then the actual models we built
for model_name in [ "ridge_air_only", "ridge_air_context", "hgb_air_context" ]:
    estimator = air_water_model_store.get( model_name )
    feature_cols = air_water_feature_store.get( model_name, [ ] )
    fill_values = air_water_fill_store.get( model_name, pd.Series( dtype = "float64" ) )
    pred_col = air_water_pred_col_store.get( model_name )

    # if we don't have an estimator, or any features, or a pred col, then skip this one
    if estimator is None or len( feature_cols ) == 0 or pred_col is None:
        continue

    # make predictions on the holdout set, but only if we have some rows and features to predict on
    if len( air_water_holdout_model ) > 0:
        X_holdout_model = air_water_holdout_model[ feature_cols ].fillna( fill_values )
        air_water_holdout_model[ pred_col ] = estimator.predict( X_holdout_model )
        append_air_water_score( 
            air_water_score_rows, 
            model_name, 
            y_holdout, 
            air_water_holdout_model[ pred_col ]
        )

if len( air_water_holdout_model ) == 0:
    print( "stage 6 note: no holdout rows survived the current filters, so the score tables will be empty" )
```

### Stage 7: Water Response Models

### 7.0 Build Daily Response Base

First we rebuild a daily table with the stage 6 water-temperature prediction attached. That predicted warming signal becomes one of the main drivers for the salinity, oxygen, and pH response models.


```python
# and now, let's prep the station-year response table
air_source_7 = add_air_water_features( daily_air_final.merge( station_reference_means, on = [ "region", "station" ], how = "left" ) )
water_source_7 = daily_water_final

response_base_target_candidates = [ "delta_salinity", "delta_oxygen", "delta_ph" ]
response_base_targets = [ col for col in response_base_target_candidates if col in water_source_7.columns ]

water_keep_cols_7 = [ "region", "station", "date" ] + response_base_targets
for col in [ "oxygen", "water_temp_baseline", "salinity_baseline", "oxygen_baseline", "ph_baseline", "depth_baseline" ]:
    if col in water_source_7.columns:
        water_keep_cols_7.append( col )
water_keep_cols_7 = list( dict.fromkeys( water_keep_cols_7 ) )

response_daily = air_source_7.merge( 
    water_source_7[ water_keep_cols_7 ],
    on = [ "region", "station", "date" ],
    how = "inner",
)
response_daily = response_daily.merge( air_water_station_context, on = [ "region", "station" ], how = "left" )

response_cluster_lookup = station_baseline[ [ "region", "station", "cluster" ] ]
response_cluster_lookup[ "cluster_code" ] = pd.to_numeric( response_cluster_lookup[ "cluster" ], errors = "coerce" )
response_cluster_lookup = response_cluster_lookup.drop( columns = [ "cluster" ] )
response_daily = response_daily.merge( response_cluster_lookup, on = [ "region", "station" ], how = "left" )
response_daily = add_air_water_features( response_daily )
```

### 7.0 Attach Stage 6 Temperature Signal

The response models should not peek at observed future water temperature. So this cell uses the selected stage 6 model to estimate `delta_water_temp_pred_p6` from air and station features.


```python
air_water_feature_cols_for_response = list( air_water_feature_cols )
for col in air_water_feature_cols_for_response:
    if col not in response_daily.columns:
        response_daily[ col ] = np.nan

air_water_fill_values_for_response = air_water_fill_values.reindex( air_water_feature_cols_for_response )
air_water_response_X = response_daily[ air_water_feature_cols_for_response ].fillna( air_water_fill_values_for_response )
response_daily[ "delta_water_temp_pred_p6" ] = air_water_model.predict( air_water_response_X )

response_daily[ "year" ] = pd.to_datetime( response_daily[ "date" ], errors = "coerce" ).dt.year
response_daily[ "month_num" ] = pd.to_datetime( response_daily[ "date" ], errors = "coerce" ).dt.month
response_daily[ "is_warm_season" ] = response_daily[ "month_num" ].between( 6, 9 )
```

### 7.0 Roll Daily Responses To Station Years

Now the daily table gets compressed into station-year summaries. Annual means are used for broad chemistry, while warm-season summaries are kept for oxygen stress.


```python
agg_feature_cols = [ col for col in [ "delta_water_temp_pred_p6", "air_temp", "precip", "wind_speed", "solar" ] if col in response_daily.columns ]
baseline_daily_feature_cols = [ col for col in [ "water_temp_baseline", "salinity_baseline", "oxygen_baseline", "ph_baseline", "depth_baseline" ] if col in response_daily.columns ]

response_annual_agg = { "n_days_annual": ( "date", "size" ) }
for col in agg_feature_cols:
    response_annual_agg[ f"{ col }_mean" ] = ( col, "mean" )
    response_annual_agg[ f"{ col }_std" ] = ( col, "std" )
for col in baseline_daily_feature_cols:
    response_annual_agg[ f"{ col }_annual_mean" ] = ( col, "mean" )
for target in response_base_targets:
    response_annual_agg[ f"{ target }_annual_mean" ] = ( target, "mean" )

response_station_year = ( 
    response_daily
    .groupby( [ "region", "station", "year" ], as_index = False )
    .agg( **response_annual_agg )
)

response_warm_daily = response_daily.loc[ response_daily[ "is_warm_season" ] ]
response_warm_agg = { "n_days_warm": ( "date", "size" ) }
for col in agg_feature_cols:
    response_warm_agg[ f"{ col }_warm_mean" ] = ( col, "mean" )
    response_warm_agg[ f"{ col }_warm_std" ] = ( col, "std" )
for target in response_base_targets:
    response_warm_agg[ f"{ target }_warm_mean" ] = ( target, "mean" )

response_warm_station_year = ( 
    response_warm_daily
    .groupby( [ "region", "station", "year" ], as_index = False )
    .agg( **response_warm_agg )
)
response_station_year = response_station_year.merge( response_warm_station_year, on = [ "region", "station", "year" ], how = "left" )
```

### 7.0 Add Oxygen Stress And Coverage Flags

Oxygen risk is not only about the mean. This cell adds warm-season lows and day-share targets, then brings in the stage 5 station-year quality flags so weak years do not define the models.


```python
def warm_quantile( values, q ):
    valid = pd.Series( values ).dropna( )
    return float( valid.quantile( q ) ) if len( valid ) > 0 else np.nan


def warm_share_below( values, threshold ):
    valid = pd.Series( values ).dropna( )
    return float( ( valid < threshold ).mean( ) ) if len( valid ) > 0 else np.nan


if "oxygen" in response_warm_daily.columns:
    response_oxygen_stress = ( 
        response_warm_daily
        .groupby( [ "region", "station", "year" ], as_index = False )
        .agg( 
            oxygen_warm_min = ( "oxygen", "min" ),
            oxygen_warm_p10 = ( "oxygen", lambda values: warm_quantile( values, 0.10 ) ),
            oxygen_warm_hypoxic_day_share = ( "oxygen", lambda values: warm_share_below( values, 2.0 ) ),
            oxygen_warm_below_optimal_day_share = ( "oxygen", lambda values: warm_share_below( values, 5.0 ) )
        )
    )
    response_station_year = response_station_year.merge( response_oxygen_stress, on = [ "region", "station", "year" ], how = "left" )

response_station_year = response_station_year.merge( 
    station_year_quality[ 
        [ 
            "region",
            "station",
            "year",
            "has_model_year",
            "has_warm_oxygen_year",
            "n_temp_days",
            "n_temp_months",
            "n_warm_oxygen_days",
            "n_warm_oxygen_months"
        ]
    ],
    on = [ "region", "station", "year" ],
    how = "left"
)
response_station_year[ "has_model_year" ] = response_station_year[ "has_model_year" ].fillna( False ).astype( bool )
response_station_year[ "has_warm_oxygen_year" ] = response_station_year[ "has_warm_oxygen_year" ].fillna( False ).astype( bool )
```

### 7.0 Response Targets

This is the contract for stage 7: each named target points to a real column, a feature set, and the coverage flag needed before a station-year can train or test that target.


```python
response_target_col_map = { 
    "delta_salinity": "delta_salinity_annual_mean" if "delta_salinity_annual_mean" in response_station_year_train.columns else None,
    "delta_oxygen": "delta_oxygen_warm_mean" if "delta_oxygen_warm_mean" in response_station_year_train.columns else ( "delta_oxygen_annual_mean" if "delta_oxygen_annual_mean" in response_station_year_train.columns else None ),
    "oxygen_warm_min": "oxygen_warm_min" if "oxygen_warm_min" in response_station_year_train.columns else None,
    "oxygen_warm_p10": "oxygen_warm_p10" if "oxygen_warm_p10" in response_station_year_train.columns else None,
    "oxygen_warm_hypoxic_day_share": "oxygen_warm_hypoxic_day_share" if "oxygen_warm_hypoxic_day_share" in response_station_year_train.columns else None,
    "oxygen_warm_below_optimal_day_share": "oxygen_warm_below_optimal_day_share" if "oxygen_warm_below_optimal_day_share" in response_station_year_train.columns else None,
    "delta_ph": "delta_ph_annual_mean" if "delta_ph_annual_mean" in response_station_year_train.columns else None
}
response_target_feature_map = { 
    "delta_salinity": list( annual_response_features ),
    "delta_oxygen": list( warm_response_features if use_warm_oxygen_features else annual_response_features ),
    "oxygen_warm_min": list( warm_response_features if use_warm_oxygen_features else annual_response_features ),
    "oxygen_warm_p10": list( warm_response_features if use_warm_oxygen_features else annual_response_features ),
    "oxygen_warm_hypoxic_day_share": list( warm_response_features if use_warm_oxygen_features else annual_response_features ),
    "oxygen_warm_below_optimal_day_share": list( warm_response_features if use_warm_oxygen_features else annual_response_features ),
    "delta_ph": list( annual_response_features )
}
response_target_year_flag_map = { 
    "delta_salinity": "has_model_year",
    "delta_oxygen": "has_warm_oxygen_year" if use_warm_oxygen_features else "has_model_year",
    "oxygen_warm_min": "has_warm_oxygen_year" if use_warm_oxygen_features else "has_model_year",
    "oxygen_warm_p10": "has_warm_oxygen_year" if use_warm_oxygen_features else "has_model_year",
    "oxygen_warm_hypoxic_day_share": "has_warm_oxygen_year" if use_warm_oxygen_features else "has_model_year",
    "oxygen_warm_below_optimal_day_share": "has_warm_oxygen_year" if use_warm_oxygen_features else "has_model_year",
    "delta_ph": "has_model_year"
}
response_target_window_label_map = { 
    "delta_salinity": "annual mean",
    "delta_oxygen": "warm-season mean" if response_target_col_map.get( "delta_oxygen" ) == "delta_oxygen_warm_mean" else "annual mean",
    "oxygen_warm_min": "warm-season minimum",
    "oxygen_warm_p10": "warm-season 10th percentile",
    "oxygen_warm_hypoxic_day_share": "warm-season share < 2 mg/L",
    "oxygen_warm_below_optimal_day_share": "warm-season share < 5 mg/L",
    "delta_ph": "annual mean"
}
response_share_targets = [ "oxygen_warm_hypoxic_day_share", "oxygen_warm_below_optimal_day_share" ]
response_targets = [ 
    target
    for target in [ 
        "delta_salinity",
        "delta_oxygen",
        "oxygen_warm_min",
        "oxygen_warm_p10",
        "oxygen_warm_hypoxic_day_share",
        "oxygen_warm_below_optimal_day_share",
        "delta_ph"
    ]
    if response_target_col_map.get( target ) is not None and len( response_target_feature_map.get( target, [ ] ) ) > 0
]
```

### 7.1 Prediction Wrapper

This tiny wrapper lets calibrated and uncalibrated models behave the same way later. For share targets, it also keeps predictions inside the possible 0 to 1 range.


```python
class ResponseTargetPredictor:


    def __init__( self, base_model, target_name, calibrator = None ):
        self.base_model = base_model
        self.target_name = target_name
        self.calibrator = calibrator


    def predict( self, X ):
        pred = np.asarray( self.base_model.predict( X ), dtype = "float64" )

        if self.calibrator is not None:
            pred = self.calibrator.predict( pred.reshape( -1, 1 ) )

        if self.target_name in response_share_targets:
            pred = np.clip( pred, 0.0, 1.0 )

        return pred
```

### 7.1 Scoring And Weighting Helpers

The score helper records the same metrics for every target. The weighting helper nudges models to care more about the tails: low oxygen values, high stress shares, and large chemistry deviations.


```python
def append_response_score( score_rows, target, target_window, model_name, y_true, y_pred, n_train, n_holdout, n_features, coverage_flag, target_col ):
    score_rows.append( 
        { 
            "target": target,
            "target_window": target_window,
            "model": model_name,
            "mae": float( mean_absolute_error( y_true, y_pred ) ),
            "rmse": float( mean_squared_error( y_true, y_pred ) ** 0.5 ),
            "r2": float( r2_score( y_true, y_pred ) ),
            "bias": float( np.mean( np.asarray( y_pred ) - np.asarray( y_true ) ) ),
            "n_train": int( n_train ),
            "n_holdout": int( n_holdout ),
            "n_features": int( n_features ),
            "coverage_flag": coverage_flag,
            "target_col": target_col,
        }
    )


def build_response_sample_weights( y, target ):
    y = pd.Series( y, copy = False ).astype( "float64" )
    weights = pd.Series( 1.0, index = y.index, dtype = "float64" )
    valid = y.dropna( )

    if len( valid ) < 8:
        return weights

    if target in [ "delta_oxygen", "oxygen_warm_min", "oxygen_warm_p10" ]:
        q25 = float( valid.quantile( 0.25 ) )
        q10 = float( valid.quantile( 0.10 ) )
        weights.loc[ y <= q25 ] = 2.0
        weights.loc[ y <= q10 ] = 3.0

    elif target in response_share_targets:
        q75 = float( valid.quantile( 0.75 ) )
        q90 = float( valid.quantile( 0.90 ) )
        weights.loc[ y >= q75 ] = 2.0
        weights.loc[ y >= q90 ] = 3.0

    else:
        q20 = float( valid.quantile( 0.20 ) )
        q80 = float( valid.quantile( 0.80 ) )
        weights.loc[ ( y <= q20 ) | ( y >= q80 ) ] = 1.75

    return weights
```

### 7.1 Out-Of-Fold Calibration Helper

Calibration uses held-out station groups inside the training set. That keeps the calibration step from simply memorizing the same station-year rows it just fit.


```python
def build_response_group_oof_predictions( estimator, X, y, groups, sample_weight = None ):
    groups = pd.Series( groups, index = X.index ).astype( str )
    sample_weight = pd.Series( sample_weight, index = X.index ).astype( "float64" ) if sample_weight is not None else None
    unique_groups = groups.dropna( ).unique( )

    if len( unique_groups ) < 2:
        fitted_estimator = clone( estimator )
        if sample_weight is None:
            fitted_estimator.fit( X, y )

        else:
            fitted_estimator.fit( X, y, sample_weight = sample_weight )
        return pd.Series( fitted_estimator.predict( X ), index = X.index, dtype = "float64" )

    splitter = GroupKFold( n_splits = min( 5, len( unique_groups ) ) )
    oof_pred = pd.Series( index = X.index, dtype = "float64" )

    for fit_idx, valid_idx in splitter.split( X, y, groups ):
        fold_model = clone( estimator )
        if sample_weight is None:
            fold_model.fit( X.iloc[ fit_idx ], y.iloc[ fit_idx ] )

        else:
            fold_model.fit( X.iloc[ fit_idx ], y.iloc[ fit_idx ], sample_weight = sample_weight.iloc[ fit_idx ] )
        oof_pred.iloc[ valid_idx ] = fold_model.predict( X.iloc[ valid_idx ] )

    if oof_pred.isna( ).any( ):
        fallback_model = clone( estimator )
        if sample_weight is None:
            fallback_model.fit( X, y )

        else:
            fallback_model.fit( X, y, sample_weight = sample_weight )
        missing_mask = oof_pred.isna( )
        oof_pred.loc[ missing_mask ] = fallback_model.predict( X.loc[ missing_mask ] )

    return oof_pred
```

### 7.1 Fit Response Models

This loop is dense, but the idea is simple: for each target, build clean train/holdout matrices, score a naive median baseline, try a small candidate set, and store the best learned model for scenarios.


```python
response_score_rows = [ ]
response_best_rows = [ ]
response_model_store = { }
response_feature_store = { }
response_fill_store = { }

# one pass per target keeps the tables comparable, even though each target has its own coverage rule
for target in response_targets:
    target_col = response_target_col_map.get( target )
    feature_cols_t = response_target_feature_map.get( target, [ ] )
    year_flag_col = response_target_year_flag_map.get( target, "has_model_year" )

    train_t = response_station_year_train.loc[ response_station_year_train[ year_flag_col ].fillna( False ).astype( bool ) ]
    holdout_t = response_station_year_holdout.loc[ response_station_year_holdout[ year_flag_col ].fillna( False ).astype( bool ) ]
    train_t = train_t.dropna( subset = [ target_col ] )
    holdout_t = holdout_t.dropna( subset = [ target_col ] )

    # if a target has no usable train/holdout rows, skip it quietly and keep the rest moving
    if len( train_t ) == 0 or len( holdout_t ) == 0:
        continue

    y_train_t = train_t[ target_col ]
    y_holdout_t = holdout_t[ target_col ]
    X_train_t = train_t[ feature_cols_t ].apply( pd.to_numeric, errors = "coerce" )
    X_holdout_t = holdout_t[ feature_cols_t ].apply( pd.to_numeric, errors = "coerce" )

    usable_feature_cols_t = X_train_t.columns[ X_train_t.notna( ).any( axis = 0 ) ].tolist( )
    if len( usable_feature_cols_t ) == 0:
        continue

    X_train_t = X_train_t[ usable_feature_cols_t ]
    X_holdout_t = X_holdout_t.reindex( columns = usable_feature_cols_t )
    fill_values_t = X_train_t.median( numeric_only = True ).reindex( usable_feature_cols_t ).fillna( 0.0 )
    X_train_t = X_train_t.fillna( fill_values_t )
    X_holdout_t = X_holdout_t.fillna( fill_values_t )
    feature_cols_t = usable_feature_cols_t
    station_groups_t = train_t[ "region" ].astype( str ) + " | " + train_t[ "station" ].astype( str )
    sample_weight_t = build_response_sample_weights( y_train_t, target )

    # global median is the boring baseline every learned model has to beat
    pred_naive = np.full( len( y_holdout_t ), float( y_train_t.median( ) ) )
    pred_naive = np.clip( pred_naive, 0.0, 1.0 ) if target in response_share_targets else pred_naive
    append_response_score( 
        response_score_rows,
        target,
        response_target_window_label_map.get( target, "annual mean" ),
        "naive_global_median",
        y_holdout_t,
        pred_naive,
        len( train_t ),
        len( holdout_t ),
        len( feature_cols_t ),
        year_flag_col,
        target_col
    )

    # same short candidate list for each response target: ridge, weighted ridge, HGB, weighted HGB
    model_candidates_t = { }

    ridge_t = Ridge( alpha = 1.0 )
    ridge_t.fit( X_train_t, y_train_t )
    model_candidates_t[ "ridge_response_climate" ] = ResponseTargetPredictor( ridge_t, target )

    ridge_weighted_t = Ridge( alpha = 1.0 )
    ridge_weighted_t.fit( X_train_t, y_train_t, sample_weight = sample_weight_t )
    ridge_oof_t = build_response_group_oof_predictions( Ridge( alpha = 1.0 ), X_train_t, y_train_t, station_groups_t, sample_weight = sample_weight_t )
    ridge_calibrator_t = LinearRegression( )
    ridge_calibrator_t.fit( np.asarray( ridge_oof_t ).reshape( -1, 1 ), y_train_t, sample_weight = sample_weight_t )
    model_candidates_t[ "ridge_response_weighted_calibrated" ] = ResponseTargetPredictor( ridge_weighted_t, target, ridge_calibrator_t )

    hgb_t = HistGradientBoostingRegressor( 
        learning_rate = 0.05,
        max_depth = 4,
        max_iter = 300,
        min_samples_leaf = 20,
        random_state = 42
    )
    hgb_t.fit( X_train_t, y_train_t )
    model_candidates_t[ "hgb_response_climate" ] = ResponseTargetPredictor( hgb_t, target )

    hgb_weighted_t = HistGradientBoostingRegressor( 
        learning_rate = 0.05,
        max_depth = 4,
        max_iter = 300,
        min_samples_leaf = 20,
        random_state = 42
    )
    hgb_weighted_t.fit( X_train_t, y_train_t, sample_weight = sample_weight_t )
    hgb_oof_t = build_response_group_oof_predictions( 
        HistGradientBoostingRegressor( 
            learning_rate = 0.05,
            max_depth = 4,
            max_iter = 300,
            min_samples_leaf = 20,
            random_state = 42
        ),
        X_train_t,
        y_train_t,
        station_groups_t,
        sample_weight = sample_weight_t
    )
    hgb_calibrator_t = LinearRegression( )
    hgb_calibrator_t.fit( np.asarray( hgb_oof_t ).reshape( -1, 1 ), y_train_t, sample_weight = sample_weight_t )
    model_candidates_t[ "hgb_response_weighted_calibrated" ] = ResponseTargetPredictor( hgb_weighted_t, target, hgb_calibrator_t )

    for model_name, model_obj in model_candidates_t.items( ):
        pred_t = model_obj.predict( X_holdout_t )
        append_response_score( 
            response_score_rows,
            target,
            response_target_window_label_map.get( target, "annual mean" ),
            model_name,
            y_holdout_t,
            pred_t,
            len( train_t ),
            len( holdout_t ),
            len( feature_cols_t ),
            year_flag_col,
            target_col
        )

    target_scores = pd.DataFrame( response_score_rows )
    target_scores = target_scores.loc[ target_scores[ "target" ] == target ].sort_values( "rmse", ascending = True )
    target_scores_non_naive = target_scores.loc[ target_scores[ "model" ] != "naive_global_median" ]
    best_row_for_scenarios = target_scores_non_naive.iloc[ 0 ] if len( target_scores_non_naive ) > 0 else target_scores.iloc[ 0 ]

    # carry forward the best learned model, not the naive baseline, for scenario work
    best_model_name = str( best_row_for_scenarios[ "model" ] )
    response_model_store[ target ] = model_candidates_t[ best_model_name ]
    response_feature_store[ target ] = feature_cols_t
    response_fill_store[ target ] = fill_values_t
    response_best_rows.append( 
        { 
            "target": target,
            "target_window": response_target_window_label_map.get( target, "annual mean" ),
            "best_model_for_scenarios": best_model_name,
            "best_rmse_for_scenarios": float( best_row_for_scenarios[ "rmse" ] ),
            "best_mae_for_scenarios": float( best_row_for_scenarios[ "mae" ] ),
            "best_r2_for_scenarios": float( best_row_for_scenarios[ "r2" ] ),
            "best_bias_for_scenarios": float( best_row_for_scenarios[ "bias" ] ),
            "n_train": int( len( train_t ) ),
            "n_holdout": int( len( holdout_t ) ),
            "n_features": int( len( feature_cols_t ) ),
            "coverage_flag": year_flag_col,
            "target_col": target_col
        }
    )
```

### 9.0 Forward-Projection Setup

These knobs control the case-study forecast. By default, the notebook focuses on a few holdout transition stations in Chesapeake regions. If a side-script-built `phase9.scenario.paths.csv` is available beside the notebook, Stage 9 uses those scenario paths; otherwise it uses simple demo ramps.


```python
# forward-projection setup knobs
forecast_case_count = 5
forecast_case_pool_mode = "holdout_transition"
forecast_case_region_codes = [ "cbm", "cbv" ]
forecast_case_station_keys = None
forecast_rolling_window_years = 5
scenario_path_csv = script_dir / "phase9.scenario.paths.csv"

# stage 9 wants a display-friendly station table; older cells only build `station_baseline`
station_baseline_display = station_baseline.copy( )
if "region_name" not in station_baseline_display.columns:
    station_baseline_display[ "region_name" ] = station_baseline_display[ "region" ]

if "station_name" not in station_baseline_display.columns:
    station_baseline_display[ "station_name" ] = station_baseline_display[ "station" ]

# for reproducibility, we'll select a fixed set of station-year cases 
# to run through the forward projection and scenario analysis; 
# these will be selected from the holdout_transition pool, 
# which is the most recent data and should be most relevant to near-term forecasting


def select_forecast_case_studies( case_pool, n_cases = 5 ):
    case_pool = case_pool.copy( )
    if len( case_pool ) == 0:
        out = case_pool.head( 0 ).copy( )
        out[ "case_id" ] = pd.Series( dtype = "object" )
        return out

    case_pool = case_pool.sort_values( [ "event_date", "region", "station" ], na_position = "last" ).reset_index( drop = True )
    selected_rows = [ ]
    used_keys = set( )

    # first pass: one case per cluster, to maximize diversity
    for cluster_id in case_pool[ "cluster" ].dropna( ).astype( int ).sort_values( ).unique( ):
        cluster_slice = case_pool.loc[ case_pool[ "cluster" ].astype( "Int64" ) == int( cluster_id ) ]
        if len( cluster_slice ) == 0:
            continue

        row = cluster_slice.iloc[ 0 ]
        key = ( str( row[ "region" ] ), str( row[ "station" ] ) )
        if key in used_keys:
            continue

        selected_rows.append( row )
        used_keys.add( key )
        if len( selected_rows ) >= n_cases:
            break

    # second pass: fill remaining slots with most recent cases, 
    # skipping any stations we've already selected
    if len( selected_rows ) < n_cases:
        for row in case_pool.itertuples( index = False ):
            key = ( str( row.region ), str( row.station ) )
            if key in used_keys:
                continue

            selected_rows.append( pd.Series( row._asdict( ) ) )
            used_keys.add( key )
            if len( selected_rows ) >= n_cases:
                break

    # final sanity check to ensure we didn't accidentally select duplicate station-years
    # which would cause problems for the forward projection code
    out = pd.DataFrame( selected_rows ).reset_index( drop = True )
    out[ "case_id" ] = [ f"case_{ idx + 1 }" for idx in range( len( out ) ) ]
    return out
```

### 9.0 Station-Key Normalizer

This helper lets us pass case-study stations as dictionaries or `( region, station )` pairs without rewriting the forecast code.


```python
# helper to take the various ways users might specify station keys 
# for forecasting and turn them into a consistent dataframe format


def normalize_forecast_station_keys( station_keys ):
    if station_keys is None:
        return None

    rows = [ ]
    # we allow users to specify station keys as dicts with "region" and "station" keys, 
    # this loop normalizes those into a consistent format
    for idx, item in enumerate( station_keys ):
        if isinstance( item, dict ):
            region = item.get( "region" )
            station = item.get( "station" )

        elif isinstance( item, ( list, tuple ) ) and len( item ) >= 2:
            region, station = item[ 0 ], item[ 1 ]

        else:
            continue

        if pd.isna( region ) or pd.isna( station ):
            continue

        rows.append( 
            { 
                "region": str( region ).strip( ),
                "station": str( station ).strip( ),
                "forecast_case_order": idx
            }
        )

    return pd.DataFrame( rows ).drop_duplicates( subset = [ "region", "station" ] ).reset_index( drop = True ) if rows else None
```

### 9.0 Case-Study Pool Helper

The case-study pool can come from holdout transitions, training transitions, all transitions, or a hand-picked station list. The rest of stage 9 works the same either way.


```python
# this is the main helper function that takes the various pieces we've built and selects the station-year cases


def build_forecast_case_pool( case_pool_mode, case_station_keys = None ):

    # start with a master list of all stations and their metadata
    station_meta = station_baseline_display[ 
        [ "region", "station", "region_name", "station_name", "cluster", "cluster_label", "cluster_name" ]
    ].drop_duplicates( )

    # we can only forecast for stations that have both air and water data, 
    # since the air-water model is a key part of the projection
    available_keys = ( 
        daily_air_final[ [ "region", "station" ] ]
        .drop_duplicates( )
        .merge( daily_water_final[ [ "region", "station" ] ].drop_duplicates( ), on = [ "region", "station" ], how = "inner" )
    )

    # the transition pool is the set of stations that have a clear "first event" date, 
    # which is important for the forward projection code to run properly
    transition_pool = ( 
        first_event
        .merge( station_meta, on = [ "region", "station" ], how = "left" )
        .merge( final_flagged_stations.assign( is_holdout = True ), on = [ "region", "station" ], how = "left" )
        .merge( available_keys.assign( has_history = True ), on = [ "region", "station" ], how = "inner" )
    )
    transition_pool[ "is_holdout" ] = transition_pool[ "is_holdout" ].fillna( False ) if "is_holdout" in transition_pool.columns else False
    transition_pool[ "is_train" ] = ~transition_pool[ "is_holdout" ]

    # the available pool is the set of stations that have air and water data
    # regardless of whether they have a clear first event date
    available_pool = ( 
        available_keys
        .merge( station_meta, on = [ "region", "station" ], how = "left" )
        .merge( final_flagged_stations.assign( is_holdout = True ), on = [ "region", "station" ], how = "left" )
    )
    available_pool[ "event_date" ] = pd.NaT
    available_pool[ "is_holdout" ] = available_pool[ "is_holdout" ].fillna( False )
    available_pool[ "is_train" ] = ~available_pool[ "is_holdout" ]

    # now we can select the case pool based on the requested mode
    if case_pool_mode == "holdout_transition":
        case_pool = transition_pool.loc[ transition_pool[ "is_holdout" ] ]

    elif case_pool_mode == "train_transition":
        case_pool = transition_pool.loc[ transition_pool[ "is_train" ] ]

    elif case_pool_mode == "all_transition":
        case_pool = transition_pool

    # if the user requested a specific list of station keys
    # we can use that to filter down the station pool

    elif case_pool_mode == "station_list":
        station_keys_frame = normalize_forecast_station_keys( case_station_keys )
        if station_keys_frame is None or len( station_keys_frame ) == 0:
            raise ValueError( "forecast_case_station_keys must be provided when forecast_case_pool_mode = 'station_list'" )

        case_pool = ( 
            station_keys_frame
            .merge( station_meta, on = [ "region", "station" ], how = "left" )
            .merge( first_event, on = [ "region", "station" ], how = "left" )
            .merge( final_flagged_stations.assign( is_holdout = True ), on = [ "region", "station" ], how = "left" )
            .merge( available_keys.assign( has_history = True ), on = [ "region", "station" ], how = "inner" )
            .sort_values( [ "forecast_case_order", "region", "station" ] )
            .reset_index( drop = True )
        )
        case_pool[ "is_holdout" ] = case_pool[ "is_holdout" ].fillna( False )
        case_pool[ "is_train" ] = ~case_pool[ "is_holdout" ]

    # if the user requested an unknown mode, raise an error

    else:
        raise ValueError( f"Unknown forecast_case_pool_mode: { case_pool_mode }" )

    # if the resulting case pool is empty, we have a few fallback options depending on what data is available
    if len( case_pool ) == 0:
        if len( transition_pool ) > 0:
            print( "stage 9 note: requested case-study pool was empty; using all transition stations instead" )
            case_pool = transition_pool

        else:
            print( "stage 9 note: no transition case-study rows found; using available stations instead" )
            case_pool = available_pool

    return case_pool.sort_values( [ "event_date", "region", "station" ], na_position = "last" ).reset_index( drop = True )
```

### 9.0 Demo Scenario Helper

If no scenario CSV is available, this builds two simple annual ramps. These are not full climate products; they are lightweight paths for exercising the forecast machinery.


```python
def build_demo_scenario_paths( start_year = 2026, end_year = 2100 ):
    rows = [ ]
    demo_specs = [ 
        ( "ssp245_demo", 1.8, 1.00, 1.00, 1.00 ),
        ( "ssp585_demo", 3.7, 0.97, 1.02, 1.00 ),
    ]
    n_years = max( 1, end_year - start_year )

    for scenario_name, end_air_add_c, end_precip_mult, end_wind_mult, end_solar_mult in demo_specs:
        for year in range( start_year, end_year + 1 ):
            frac = ( year - start_year ) / n_years
            rows.append( 
                { 
                    "scenario": scenario_name,
                    "year": year,
                    "air_temp_add_c": float( end_air_add_c * frac ),
                    "precip_mult": float( 1.0 + ( end_precip_mult - 1.0 ) * frac ),
                    "wind_speed_mult": float( 1.0 + ( end_wind_mult - 1.0 ) * frac ),
                    "solar_mult": float( 1.0 + ( end_solar_mult - 1.0 ) * frac )
                }
            )

    return pd.DataFrame( rows )
```

### 9.0 Future Driver Rolls

Future daily drivers need the same rolling features as the historical stage 6 table. This helper rebuilds those 1-, 7-, and 28-day lagged means by station and scenario.


```python
def add_forecast_driver_rolls( frame ):
    frame = frame.sort_values( [ "region", "station", "scenario", "date" ] )

    for source_col in [ "air_temp", "wind_speed", "precip", "solar" ]:
        for window in [ 1, 7, 28 ]:
            frame[ f"{ source_col }_r{ window }d" ] = ( 
                frame
                .groupby( [ "region", "station", "scenario" ] )[ source_col ]
                .transform( lambda values: values.shift( 1 ).rolling( window = window, min_periods = 1 ).mean( ) )
            )

    frame[ "doy" ] = frame[ "date" ].dt.dayofyear
    frame[ "doy_sin" ] = np.sin( 2 * np.pi * frame[ "doy" ] / 365.25 )
    frame[ "doy_cos" ] = np.cos( 2 * np.pi * frame[ "doy" ] / 365.25 )
    return frame
```

### 9.0 Forecast Station-Year Helper

This helper mirrors the stage 7 station-year table, but for future scenario rows. Keeping the shape similar is what lets stage 7 models run on stage 9 forecasts.


```python
def build_forecast_station_year_features( forecast_daily_frame ):
    frame = forecast_daily_frame.dropna( subset = [ "date" ] )
    frame[ "date" ] = pd.to_datetime( frame[ "date" ], errors = "coerce" )
    frame[ "year" ] = frame[ "date" ].dt.year
    frame[ "month_num" ] = frame[ "date" ].dt.month
    frame[ "is_warm_season" ] = frame[ "month_num" ].between( 6, 9 )

    agg_feature_cols = [ col for col in [ "delta_water_temp_pred_p6", "air_temp", "precip", "wind_speed", "solar" ] if col in frame.columns ]
    baseline_daily_feature_cols = [ col for col in [ "water_temp_baseline", "salinity_baseline", "oxygen_baseline", "ph_baseline", "depth_baseline" ] if col in frame.columns ]

    annual_agg = { "n_days_annual": ( "date", "size" ) }
    for col in agg_feature_cols:
        annual_agg[ f"{ col }_mean" ] = ( col, "mean" )
        annual_agg[ f"{ col }_std" ] = ( col, "std" )
    for col in baseline_daily_feature_cols:
        annual_agg[ f"{ col }_annual_mean" ] = ( col, "mean" )

    station_year = ( 
        frame
        .groupby( [ "region", "station", "scenario", "year" ], as_index = False )
        .agg( **annual_agg )
    )

    warm_daily = frame.loc[ frame[ "is_warm_season" ] ]
    warm_agg = { "n_days_warm": ( "date", "size" ) }
    for col in agg_feature_cols:
        warm_agg[ f"{ col }_warm_mean" ] = ( col, "mean" )
        warm_agg[ f"{ col }_warm_std" ] = ( col, "std" )

    warm_year = ( 
        warm_daily
        .groupby( [ "region", "station", "scenario", "year" ], as_index = False )
        .agg( **warm_agg )
    )
    station_year = station_year.merge( warm_year, on = [ "region", "station", "scenario", "year" ], how = "left" )
    station_year = station_year.merge( response_station_context, on = [ "region", "station" ], how = "left" )
    station_year[ "has_model_year" ] = True
    station_year[ "has_warm_oxygen_year" ] = True
    return station_year
```

### 9.0 Regime Classification Helper

This helper maps rolling projected conditions back into the learned baseline regime space. It only classifies a row when enough regime features are available.


```python
# stage 5 named these rolling-regime centroids `new_centroids`; stage 9 keeps a clearer export name
centroids_z = new_centroids.copy( )
regime_feature_cols = list( centroids_z.columns )

# this is the main helper function that takes a station-year row 
# and classifies it into a rolling regime based on the available features


def classify_regime_row( row, feature_cols = None, min_features_required = 3 ):
    active_feature_cols = regime_feature_cols if feature_cols is None else feature_cols
    available = [ 
        col
        for col in active_feature_cols
        if ( 
            col in row.index
            and col in feature_mean.index
            and col in feature_scale.index
            and col in centroids_z.columns
            and pd.notna( row.get( col, np.nan ) )
        )
    ]

    # if we don't have enough features to make a classification, 
    # return NaN for all regime outputs but still report how many features we had available
    if len( available ) < min_features_required:
        return pd.Series( 
            { 
                "regime_roll5y": np.nan,
                "regime_dist_best": np.nan,
                "regime_margin": np.nan,
                "regime_n_features_used": len( available )
            }
        )

    # if we have enough features, we can proceed with the z-score calculation and distance-based classification
    values = pd.to_numeric( pd.Series( { col: row.get( col, np.nan ) for col in available } ), errors = "coerce" )
    means = pd.to_numeric( feature_mean[ available ], errors = "coerce" )
    scales = pd.to_numeric( feature_scale[ available ], errors = "coerce" ).replace( 0, np.nan )
    z_values = ( values - means ) / scales
    if z_values.isna( ).any( ):
        return pd.Series( 
            { 
                "regime_roll5y": np.nan,
                "regime_dist_best": np.nan,
                "regime_margin": np.nan,
                "regime_n_features_used": len( available )
            }
        )

    # compute distances to centroids using only the available features
    centroid_slice = centroids_z[ available ]
    centroid_array = np.asarray( centroid_slice.to_numpy( ), dtype = "float64" )
    z_array = np.asarray( z_values.to_numpy( ), dtype = "float64" )
    distances = np.sqrt( np.sum( ( centroid_array - z_array[ None, : ] ) ** 2, axis = 1 ) )

    # the best cluster is the one with the smallest distance
    # the margin is the difference between the best and second-best distances
    best_idx = int( np.argmin( distances ) )
    best_cluster = int( centroid_slice.index[ best_idx ] )
    best_dist = float( distances[ best_idx ] )
    margin = float( np.partition( distances, 1 )[ 1 ] - best_dist ) if len( distances ) > 1 else np.nan

    return pd.Series( 
        { 
            "regime_roll5y": best_cluster,
            "regime_dist_best": best_dist,
            "regime_margin": margin,
            "regime_n_features_used": len( available )
        }
    )
```

#### 9.3 Helper: Share Below Threshold

This tiny helper converts a sequence of oxygen values into the share of valid observations below a threshold. The scan uses it for historical warm-season hypoxia summaries.


```python
def share_below( values, threshold ):
    values = pd.to_numeric( pd.Series( values ), errors = "coerce" ).dropna( )
    if len( values ) == 0:
        return np.nan

    return float( ( values < threshold ).mean( ) )



def share_between( values, lower_threshold, upper_threshold ):
    values = pd.to_numeric( pd.Series( values ), errors = "coerce" ).dropna( )
    if len( values ) == 0:
        return np.nan

    return float( ( ( values >= lower_threshold ) & ( values < upper_threshold ) ).mean( ) )
```

#### 9.3 Helper: Station Metadata For Scan

This builds the list of stations that can be projected in the broader scan. It keeps readable station labels, baseline regime labels, holdout/train flags, and optional region filtering in one place.


```python
def build_scan_station_meta( 
    station_baseline_display,
    daily_air_final,
    daily_water_final,
    first_event,
    final_flagged_stations,
    region_codes = None
):
    station_meta = ( 
        station_baseline_display[ 
            [ "region", "station", "region_name", "station_name", "cluster", "cluster_label", "cluster_name" ]
        ]
        .drop_duplicates( )
        .copy( )
    )
    available_keys = ( 
        daily_air_final[ [ "region", "station" ] ]
        .drop_duplicates( )
        .merge( daily_water_final[ [ "region", "station" ] ].drop_duplicates( ), on = [ "region", "station" ], how = "inner" )
    )
    station_meta = available_keys.merge( station_meta, on = [ "region", "station" ], how = "left" )
    station_meta = station_meta.merge( first_event[ [ "region", "station", "event_date" ] ], on = [ "region", "station" ], how = "left" )
    station_meta = station_meta.merge( final_flagged_stations.assign( is_holdout = True ), on = [ "region", "station" ], how = "left" )
    station_meta[ "is_holdout" ] = station_meta[ "is_holdout" ].fillna( False )
    station_meta[ "is_train" ] = ~station_meta[ "is_holdout" ]

    if region_codes is not None and len( region_codes ) > 0:
        region_codes_norm = { str( code ).strip( ).lower( ) for code in region_codes }
        station_meta = station_meta.loc[ station_meta[ "region" ].astype( str ).str.lower( ).isin( region_codes_norm ) ].copy( )

    else:
        region_codes_norm = None

    return station_meta.reset_index( drop = True ), region_codes_norm
```

#### 9.3 Helper: Historical Scan Anchors

This pulls historical air and water records for the scan stations and rolls them to annual history. Those observed annual rows become the pre-forecast anchor for regime paths and hypoxia comparisons.


```python
def build_scan_history_tables( daily_air_final, daily_water_final, station_keys, forecast_reference_means, hypoxia_threshold, stress_threshold = 5.0 ):
    hist_air = daily_air_final.merge( station_keys, on = [ "region", "station" ], how = "inner" ).copy( )
    hist_water = daily_water_final.merge( station_keys, on = [ "region", "station" ], how = "inner" ).copy( )
    hist_air[ "date" ] = pd.to_datetime( hist_air[ "date" ], errors = "coerce" )
    hist_water[ "date" ] = pd.to_datetime( hist_water[ "date" ], errors = "coerce" )

    hist_water_cols = [ col for col in [ "water_temp", "salinity", "oxygen", "depth" ] if col in hist_water.columns ]
    hist_air_for_join = hist_air.drop( 
        columns = [ col for col in hist_water_cols if col in hist_air.columns ],
        errors = "ignore",
    )
    hist_join = hist_air_for_join.merge( 
        hist_water[ [ "region", "station", "date" ] + hist_water_cols ],
        on = [ "region", "station", "date" ],
        how = "inner",
    )
    hist_join[ "year" ] = hist_join[ "date" ].dt.year
    hist_join[ "month_num" ] = hist_join[ "date" ].dt.month
    hist_join[ "is_warm_season" ] = hist_join[ "month_num" ].between( 6, 9 )

    history_annual_agg = { 
        "air_temp_mean": ( "air_temp", "mean" ),
        "precip_mean": ( "precip", "mean" ),
    }
    for col, out_col in [ 
        ( "water_temp", "water_temp_abs_annual_mean" ),
        ( "salinity", "salinity_abs_annual_mean" ),
        ( "oxygen", "oxygen_abs_annual_mean" ),
        ( "depth", "depth_abs_annual_mean" ),
    ]:
        if col in hist_join.columns:
            history_annual_agg[ out_col ] = ( col, "mean" )

    history_annual = ( 
        hist_join
        .groupby( [ "region", "station", "year" ], as_index = False )
        .agg( **history_annual_agg )
    )
    if "oxygen" in hist_join.columns:
        history_warm = ( 
            hist_join.loc[ hist_join[ "is_warm_season" ] ]
            .groupby( [ "region", "station", "year" ], as_index = False )
            .agg( 
                oxygen_warm_min_abs = ( "oxygen", "min" ),
                oxygen_warm_hypoxic_day_share_abs = ( "oxygen", lambda values: share_below( values, hypoxia_threshold ) ),
                oxygen_warm_stressed_day_share_abs = ( "oxygen", lambda values: share_between( values, hypoxia_threshold, stress_threshold ) )
            )
        )
        history_annual = history_annual.merge( history_warm, on = [ "region", "station", "year" ], how = "left" )

    else:
        history_annual[ "oxygen_warm_min_abs" ] = np.nan
        history_annual[ "oxygen_warm_hypoxic_day_share_abs" ] = np.nan

        history_annual[ "oxygen_warm_stressed_day_share_abs" ] = np.nan

    history_annual = history_annual.merge( forecast_reference_means, on = [ "region", "station" ], how = "left" )
    if "depth_abs_annual_mean" not in history_annual.columns and "depth_baseline" in history_annual.columns:
        history_annual[ "depth_abs_annual_mean" ] = history_annual[ "depth_baseline" ]
    history_annual[ "scenario" ] = "observed_history"
    history_annual[ "oxygen_plot_abs" ] = history_annual[ "oxygen_warm_min_abs" ]
    history_annual.loc[ history_annual[ "oxygen_plot_abs" ].isna( ), "oxygen_plot_abs" ] = history_annual.loc[ history_annual[ "oxygen_plot_abs" ].isna( ), "oxygen_abs_annual_mean" ]

    return hist_air, history_annual
```

#### 9.3 Helper: Driver Climatology

This builds each station's historical day-of-year weather pattern. The future scan starts from this climatology, then applies scenario adjustments year by year.


```python
def build_scan_driver_climatology( hist_air ):
    historical_climatology_source = hist_air.copy( )
    historical_climatology_source[ "doy" ] = historical_climatology_source[ "date" ].dt.dayofyear.clip( upper = 365 )
    station_driver_climatology = ( 
        historical_climatology_source
        .groupby( [ "region", "station", "doy" ], as_index = False )
        .agg( 
            air_temp = ( "air_temp", "mean" ),
            precip = ( "precip", "mean" ),
            wind_speed = ( "wind_speed", "mean" ),
            solar = ( "solar", "mean" )
        )
    )
    station_driver_means = ( 
        hist_air
        .groupby( [ "region", "station" ], as_index = False )
        .agg( 
            air_temp_station_mean = ( "air_temp", "mean" ),
            precip_station_mean = ( "precip", "mean" ),
            wind_speed_station_mean = ( "wind_speed", "mean" ),
            solar_station_mean = ( "solar", "mean" )
        )
    )

    return station_driver_climatology, station_driver_means
```

#### 9.3 Helper: Future Daily Drivers

This expands scan station keys across scenarios and future dates, attaches the historical climatology, applies scenario adjustments, and rebuilds the rolling driver features.


```python
def build_scan_future_daily( 
    station_keys,
    forecast_scenarios,
    forecast_dates,
    station_driver_climatology,
    station_driver_means,
    scenario_paths
):
    future_daily = ( 
        station_keys.assign( _tmp = 1 )
        .merge( pd.DataFrame( { "scenario": forecast_scenarios, "_tmp": 1 } ), on = "_tmp", how = "inner" )
        .merge( forecast_dates.assign( _tmp = 1 ), on = "_tmp", how = "inner" )
        .drop( columns = [ "_tmp" ] )
    )
    future_daily = future_daily.merge( station_driver_climatology, on = [ "region", "station", "doy" ], how = "left" )
    future_daily = future_daily.merge( station_driver_means, on = [ "region", "station" ], how = "left" )
    future_daily = future_daily.merge( scenario_paths, on = [ "scenario", "year" ], how = "left" )

    for col, station_fill_col in [ 
        ( "air_temp", "air_temp_station_mean" ),
        ( "precip", "precip_station_mean" ),
        ( "wind_speed", "wind_speed_station_mean" ),
        ( "solar", "solar_station_mean" )
    ]:
        future_daily[ col ] = future_daily[ col ].fillna( future_daily[ station_fill_col ] )

    future_daily[ "air_temp" ] = future_daily[ "air_temp" ] + future_daily[ "air_temp_add_c" ]
    future_daily[ "precip" ] = future_daily[ "precip" ] * future_daily[ "precip_mult" ]
    future_daily[ "wind_speed" ] = future_daily[ "wind_speed" ] * future_daily[ "wind_speed_mult" ]
    future_daily[ "solar" ] = future_daily[ "solar" ] * future_daily[ "solar_mult" ]

    return add_forecast_driver_rolls( future_daily )
```

#### 9.3 Helper: Apply Stage 6 And Stage 7 Models

This runs the same two-step modeling chain used by the case studies: Stage 6 predicts future water-temperature deltas, then Stage 7 predicts salinity, oxygen, pH, and oxygen-stress response targets.


```python
def apply_scan_models( 
    future_daily,
    forecast_reference_means,
    air_water_station_context,
    forecast_cluster_lookup,
    air_water_feature_cols,
    air_water_fill_values,
    air_water_model,
    response_targets,
    response_feature_store,
    response_fill_store,
    response_model_store
):
    future_daily = future_daily.merge( forecast_reference_means, on = [ "region", "station" ], how = "left" )
    future_daily = future_daily.merge( air_water_station_context, on = [ "region", "station" ], how = "left" )
    future_daily = future_daily.merge( forecast_cluster_lookup, on = [ "region", "station" ], how = "left" )
    future_daily = add_air_water_features( future_daily )

    for col in air_water_feature_cols:
        if col not in future_daily.columns:
            future_daily[ col ] = np.nan

    X_p6 = future_daily[ air_water_feature_cols ].copy( ).fillna( air_water_fill_values.reindex( air_water_feature_cols ) )
    future_daily[ "delta_water_temp_pred_p6" ] = air_water_model.predict( X_p6 )
    future_daily[ "water_temp_pred" ] = future_daily[ "water_temp_baseline" ] + future_daily[ "delta_water_temp_pred_p6" ]

    future_station_year = build_forecast_station_year_features( future_daily )
    for target in response_targets:
        model_t = response_model_store.get( target )
        feature_cols_t = response_feature_store.get( target, [ ] )
        fill_values_t = response_fill_store.get( target, pd.Series( dtype = "float64" ) )
        if model_t is None or len( feature_cols_t ) == 0:
            continue

        X_target_t = future_station_year[ feature_cols_t ].copy( ).fillna( fill_values_t.reindex( feature_cols_t ) )
        future_station_year[ f"{ target }_pred" ] = model_t.predict( X_target_t )

    future_station_year[ "water_temp_abs_annual_mean" ] = future_station_year[ "water_temp_baseline_annual_mean" ] + future_station_year[ "delta_water_temp_pred_p6_mean" ]
    future_station_year[ "salinity_abs_annual_mean" ] = future_station_year[ "salinity_baseline_annual_mean" ] + future_station_year.get( "delta_salinity_pred", 0.0 )
    future_station_year[ "oxygen_abs_annual_mean" ] = future_station_year[ "oxygen_baseline_annual_mean" ] + future_station_year.get( "delta_oxygen_pred", np.nan )
    future_station_year[ "oxygen_warm_min_abs" ] = future_station_year.get( "oxygen_warm_min_pred", np.nan )
    future_station_year[ "oxygen_warm_hypoxic_day_share_abs" ] = future_station_year.get( "oxygen_warm_hypoxic_day_share_pred", np.nan )

    stressed_below5_pred = future_station_year.get( "oxygen_warm_below_optimal_day_share_pred", pd.Series( np.nan, index = future_station_year.index ) )
    stressed_hypoxic_pred = future_station_year.get( "oxygen_warm_hypoxic_day_share_pred", pd.Series( 0.0, index = future_station_year.index ) )
    future_station_year[ "oxygen_warm_stressed_day_share_abs" ] = ( stressed_below5_pred - stressed_hypoxic_pred ).clip( lower = 0.0 )
    future_station_year[ "oxygen_plot_abs" ] = future_station_year[ "oxygen_warm_min_abs" ]
    future_station_year.loc[ future_station_year[ "oxygen_plot_abs" ].isna( ), "oxygen_plot_abs" ] = future_station_year.loc[ future_station_year[ "oxygen_plot_abs" ].isna( ), "oxygen_abs_annual_mean" ]

    return future_station_year
```

#### 9.3 Helper: Regime Paths

This stitches observed history and future forecasts together, calculates rolling five-year signatures, and classifies each row back into the learned baseline regime space.


```python
def add_scan_regime_paths( history_annual, future_station_year, forecast_scenarios, forecast_year_min, forecast_rolling_window_years ):
    metric_roll_map = { 
        "air_temp_mean": "air_temp_roll5y",
        "precip_mean": "precip_roll5y",
        "water_temp_abs_annual_mean": "water_temp_roll5y",
        "salinity_abs_annual_mean": "salinity_roll5y",
        "oxygen_abs_annual_mean": "oxygen_roll5y",
        "oxygen_plot_abs": "oxygen_plot_roll5y",
        "oxygen_warm_hypoxic_day_share_abs": "oxygen_hypoxic_share_roll5y",

        "oxygen_warm_stressed_day_share_abs": "oxygen_stressed_share_roll5y"
    }

    history_plot = history_annual.copy( )
    for source_col, out_col in metric_roll_map.items( ):
        history_plot[ out_col ] = ( 
            history_plot
            .sort_values( [ "region", "station", "year" ] )
            .groupby( [ "region", "station" ] )[ source_col ]
            .transform( lambda values: values.rolling( window = forecast_rolling_window_years, min_periods = 3 ).mean( ) )
        )

    history_plot[ "mean_annual_water_temp" ] = history_plot[ "water_temp_roll5y" ]
    history_plot[ "mean_annual_salinity" ] = history_plot[ "salinity_roll5y" ]
    history_plot[ "mean_annual_oxygen" ] = history_plot[ "oxygen_roll5y" ]
    history_plot[ "mean_annual_depth" ] = history_plot[ "depth_baseline" ]
    history_plot = pd.concat( [ history_plot, history_plot.apply( classify_regime_row, axis = 1 ) ], axis = 1 )

    projection_paths = [ ]
    for scenario_name in forecast_scenarios:
        hist_path = history_annual.copy( )
        hist_path[ "scenario" ] = scenario_name
        future_path = future_station_year.loc[ future_station_year[ "scenario" ] == scenario_name ].copy( )
        path_frame = pd.concat( [ hist_path, future_path ], ignore_index = True, sort = False )
        path_frame = path_frame.sort_values( [ "region", "station", "year" ] ).reset_index( drop = True )

        for source_col, out_col in metric_roll_map.items( ):
            path_frame[ out_col ] = ( 
                path_frame
                .groupby( [ "region", "station" ] )[ source_col ]
                .transform( lambda values: values.rolling( window = forecast_rolling_window_years, min_periods = 3 ).mean( ) )
            )

        path_frame[ "mean_annual_water_temp" ] = path_frame[ "water_temp_roll5y" ]
        path_frame[ "mean_annual_salinity" ] = path_frame[ "salinity_roll5y" ]
        path_frame[ "mean_annual_oxygen" ] = path_frame[ "oxygen_roll5y" ]
        path_frame[ "mean_annual_depth" ] = path_frame[ "depth_baseline" ]
        path_frame = pd.concat( [ path_frame, path_frame.apply( classify_regime_row, axis = 1 ) ], axis = 1 )
        projection_paths.append( path_frame )

    projection_plot = pd.concat( projection_paths, ignore_index = True ) if len( projection_paths ) > 0 else pd.DataFrame( )
    projection_future_only = projection_plot.loc[ projection_plot[ "year" ] >= forecast_year_min ].copy( )

    return history_plot, projection_plot, projection_future_only
```

#### 9.3 Helper: Scan Summary Tables

This condenses the large projection path into the two portfolio views we actually inspect: earliest projected regime crossings and strongest projected hypoxia-risk changes.


```python
def summarize_scan_results( station_meta, history_annual, projection_future_only, forecast_year_min, top_n, hypoxia_threshold, material_share, plot_n, stress_threshold = 5.0 ):
    crossing_summary = ( 
        projection_future_only
        .loc[ 
            projection_future_only[ "regime_roll5y" ].notna( )
            & projection_future_only[ "cluster" ].notna( )
            & ( projection_future_only[ "regime_roll5y" ].astype( "Int64" ) != projection_future_only[ "cluster" ].astype( "Int64" ) )
        ]
        .groupby( [ "region", "station", "scenario" ], as_index = False )[ "year" ]
        .min( )
        .rename( columns = { "year": "first_regime_cross_year" } )
    )
    crossing_candidates = station_meta.merge( crossing_summary, on = [ "region", "station" ], how = "inner" )
    crossing_candidates[ "first_regime_cross_decade" ] = ( crossing_candidates[ "first_regime_cross_year" ] // 10 ) * 10
    crossing_candidates = crossing_candidates.sort_values( [ "scenario", "first_regime_cross_year", "region", "station" ] ).reset_index( drop = True )

    history_hypoxia = ( 
        history_annual
        .groupby( [ "region", "station" ], as_index = False )
        .agg( 
            hist_hypoxic_share_peak = ( "oxygen_warm_hypoxic_day_share_abs", "max" ),
            hist_hypoxic_share_mean = ( "oxygen_warm_hypoxic_day_share_abs", "mean" ),
            hist_warm_min_floor = ( "oxygen_warm_min_abs", "min" ),
            hist_warm_min_mean = ( "oxygen_warm_min_abs", "mean" ),
            hist_stressed_share_peak = ( "oxygen_warm_stressed_day_share_abs", "max" ),
            hist_stressed_share_mean = ( "oxygen_warm_stressed_day_share_abs", "mean" )
        )
    )
    history_hypoxia[ "hist_hypoxia_any" ] = ( 
        history_hypoxia[ "hist_hypoxic_share_peak" ].fillna( 0.0 ) > 0.0
    ) | ( 
        history_hypoxia[ "hist_warm_min_floor" ].fillna( np.inf ) < hypoxia_threshold
    )
    history_hypoxia[ "hist_stressed_any" ] = ( 
        history_hypoxia[ "hist_stressed_share_peak" ].fillna( 0.0 ) > 0.0
    ) | ( 
        history_hypoxia[ "hist_warm_min_floor" ].fillna( np.inf ) < stress_threshold
    )

    future_hypoxia = ( 
        projection_future_only
        .groupby( [ "region", "station", "scenario" ], as_index = False )
        .agg( 
            future_hypoxic_share_peak = ( "oxygen_warm_hypoxic_day_share_abs", "max" ),
            future_hypoxic_share_mean = ( "oxygen_warm_hypoxic_day_share_abs", "mean" ),
            future_warm_min_floor = ( "oxygen_warm_min_abs", "min" ),
            future_warm_min_mean = ( "oxygen_warm_min_abs", "mean" ),
            future_stressed_share_peak = ( "oxygen_warm_stressed_day_share_abs", "max" ),
            future_stressed_share_mean = ( "oxygen_warm_stressed_day_share_abs", "mean" )
        )
    )
    future_hypoxia_2090s = ( 
        projection_future_only.loc[ projection_future_only[ "year" ] >= 2090 ]
        .groupby( [ "region", "station", "scenario" ], as_index = False )
        .agg( 
            future_hypoxic_share_2090s = ( "oxygen_warm_hypoxic_day_share_abs", "mean" ),
            future_warm_min_mean_2090s = ( "oxygen_warm_min_abs", "mean" ),
            future_stressed_share_2090s = ( "oxygen_warm_stressed_day_share_abs", "mean" )
        )
    )
    hypoxia_candidates = station_meta.merge( future_hypoxia, on = [ "region", "station" ], how = "inner" )
    hypoxia_candidates = hypoxia_candidates.merge( future_hypoxia_2090s, on = [ "region", "station", "scenario" ], how = "left" )
    hypoxia_candidates = hypoxia_candidates.merge( history_hypoxia, on = [ "region", "station" ], how = "left" )
    hypoxia_candidates[ "new_hypoxia_flag" ] = ( 
        ~hypoxia_candidates[ "hist_hypoxia_any" ].fillna( False )
    ) & ( 
        ( hypoxia_candidates[ "future_hypoxic_share_peak" ].fillna( 0.0 ) >= material_share )
        | ( hypoxia_candidates[ "future_warm_min_floor" ].fillna( np.inf ) < hypoxia_threshold )
    )
    hypoxia_candidates[ "hypoxia_share_increase_peak" ] = hypoxia_candidates[ "future_hypoxic_share_peak" ].fillna( 0.0 ) - hypoxia_candidates[ "hist_hypoxic_share_peak" ].fillna( 0.0 )
    hypoxia_candidates[ "hypoxia_share_increase_2090s" ] = hypoxia_candidates[ "future_hypoxic_share_2090s" ].fillna( 0.0 ) - hypoxia_candidates[ "hist_hypoxic_share_mean" ].fillna( 0.0 )
    hypoxia_candidates[ "warm_min_drop_mean_2090s" ] = hypoxia_candidates[ "hist_warm_min_mean" ] - hypoxia_candidates[ "future_warm_min_mean_2090s" ]
    hypoxia_candidates[ "hypoxia_intensifies_flag" ] = ( 
        hypoxia_candidates[ "hypoxia_share_increase_peak" ].fillna( 0.0 ) >= material_share
    ) | ( 
        hypoxia_candidates[ "warm_min_drop_mean_2090s" ].fillna( 0.0 ) >= 0.5
    )
    hypoxia_candidates[ "new_stressed_flag" ] = ( 
        ~hypoxia_candidates[ "hist_stressed_any" ].fillna( False )
    ) & ( 
        ( hypoxia_candidates[ "future_stressed_share_peak" ].fillna( 0.0 ) >= material_share )
        | ( hypoxia_candidates[ "future_warm_min_floor" ].fillna( np.inf ) < stress_threshold )
    )
    hypoxia_candidates[ "stressed_share_increase_peak" ] = hypoxia_candidates[ "future_stressed_share_peak" ].fillna( 0.0 ) - hypoxia_candidates[ "hist_stressed_share_peak" ].fillna( 0.0 )
    hypoxia_candidates[ "stressed_share_increase_2090s" ] = hypoxia_candidates[ "future_stressed_share_2090s" ].fillna( 0.0 ) - hypoxia_candidates[ "hist_stressed_share_mean" ].fillna( 0.0 )
    hypoxia_candidates[ "stressed_intensifies_flag" ] = ( 
        hypoxia_candidates[ "stressed_share_increase_peak" ].fillna( 0.0 ) >= material_share
    ) | ( 
        hypoxia_candidates[ "warm_min_drop_mean_2090s" ].fillna( 0.0 ) >= 0.5
    )
    hypoxia_candidates = hypoxia_candidates.sort_values( 
        [ "scenario", "new_hypoxia_flag", "hypoxia_share_increase_peak", "future_hypoxic_share_peak", "future_warm_min_floor" ],
        ascending = [ True, False, False, False, True ]
    ).reset_index( drop = True )

    crossing_display = crossing_candidates[ 
        [ 
            "scenario",
            "region",
            "station",
            "station_name",
            "region_name",
            "cluster_label",
            "is_holdout",
            "first_regime_cross_year",
            "first_regime_cross_decade"
        ]
    ].head( top_n )
    hypoxia_display = hypoxia_candidates[ 
        [ 
            "scenario",
            "region",
            "station",
            "station_name",
            "region_name",
            "cluster_label",
            "is_holdout",
            "new_hypoxia_flag",
            "hypoxia_intensifies_flag",
            "hist_hypoxic_share_peak",
            "future_hypoxic_share_peak",
            "hypoxia_share_increase_peak",
            "hist_warm_min_floor",
            "future_warm_min_floor"
        ]
    ].head( top_n )

    stress_candidates = hypoxia_candidates.sort_values( 
        [ "scenario", "new_stressed_flag", "stressed_share_increase_peak", "future_stressed_share_peak", "future_warm_min_floor" ],
        ascending = [ True, False, False, False, True ]
    ).reset_index( drop = True )
    stress_display = stress_candidates[ 
        [ 
            "scenario",
            "region",
            "station",
            "station_name",
            "region_name",
            "cluster_label",
            "is_holdout",
            "new_stressed_flag",
            "stressed_intensifies_flag",
            "hist_stressed_share_peak",
            "future_stressed_share_peak",
            "stressed_share_increase_peak",
            "hist_warm_min_floor",
            "future_warm_min_floor"
        ]
    ].head( top_n )

    plot_station_keys = pd.concat( 
        [ 
            crossing_display[ [ "region", "station" ] ],
            hypoxia_display[ [ "region", "station" ] ],
            stress_display[ [ "region", "station" ] ]
        ],
        ignore_index = True
    ).drop_duplicates( ).head( plot_n )
    plot_studies = station_meta.merge( plot_station_keys, on = [ "region", "station" ], how = "inner" )

    return crossing_candidates, hypoxia_candidates, stress_candidates, crossing_display, hypoxia_display, stress_display, plot_studies
```

#### 9.3 Helper: Scan Orchestrator

This final helper calls the smaller scan pieces in order and returns one dictionary of scan outputs. Keeping the orchestration separate makes the run cell below much easier to read.


```python
def run_stage9_station_scan( 
    *,
    station_baseline_display,
    daily_air_final,
    daily_water_final,
    first_event,
    final_flagged_stations,
    scenario_paths,
    forecast_scenarios,
    forecast_year_min,
    forecast_year_max,
    forecast_dates,
    forecast_reference_means,
    forecast_cluster_lookup,
    forecast_rolling_window_years,
    air_water_station_context,
    air_water_feature_cols,
    air_water_fill_values,
    air_water_model,
    response_targets,
    response_feature_store,
    response_fill_store,
    response_model_store,
    region_codes = None,
    top_n = 12,
    plot_n = 6,
    hypoxia_threshold = 2.0,
    material_share = 0.05,
    progress_every = 20,
    **unused
):
    station_meta, region_codes_norm = build_scan_station_meta( 
        station_baseline_display = station_baseline_display,
        daily_air_final = daily_air_final,
        daily_water_final = daily_water_final,
        first_event = first_event,
        final_flagged_stations = final_flagged_stations,
        region_codes = region_codes
    )
    station_keys = station_meta[ [ "region", "station" ] ].drop_duplicates( ).reset_index( drop = True )

    print( f"forecast scan stations to project: { len( station_keys ) }" )
    hist_air, history_annual = build_scan_history_tables( 
        daily_air_final = daily_air_final,
        daily_water_final = daily_water_final,
        station_keys = station_keys,
        forecast_reference_means = forecast_reference_means,
        hypoxia_threshold = hypoxia_threshold,
        stress_threshold = 5.0
    )
    history_annual = history_annual.merge( station_meta, on = [ "region", "station" ], how = "left" )

    station_driver_climatology, station_driver_means = build_scan_driver_climatology( hist_air )
    future_daily = build_scan_future_daily( 
        station_keys = station_keys,
        forecast_scenarios = forecast_scenarios,
        forecast_dates = forecast_dates,
        station_driver_climatology = station_driver_climatology,
        station_driver_means = station_driver_means,
        scenario_paths = scenario_paths
    )
    future_station_year = apply_scan_models( 
        future_daily = future_daily,
        forecast_reference_means = forecast_reference_means,
        air_water_station_context = air_water_station_context,
        forecast_cluster_lookup = forecast_cluster_lookup,
        air_water_feature_cols = air_water_feature_cols,
        air_water_fill_values = air_water_fill_values,
        air_water_model = air_water_model,
        response_targets = response_targets,
        response_feature_store = response_feature_store,
        response_fill_store = response_fill_store,
        response_model_store = response_model_store
    )
    future_station_year = future_station_year.merge( station_meta, on = [ "region", "station" ], how = "left" )

    history_plot, projection_plot, projection_future_only = add_scan_regime_paths( 
        history_annual = history_annual,
        future_station_year = future_station_year,
        forecast_scenarios = forecast_scenarios,
        forecast_year_min = forecast_year_min,
        forecast_rolling_window_years = forecast_rolling_window_years
    )
    crossing_candidates, hypoxia_candidates, stress_candidates, crossing_display, hypoxia_display, stress_display, plot_studies = summarize_scan_results( 
        station_meta = station_meta,
        history_annual = history_annual,
        projection_future_only = projection_future_only,
        forecast_year_min = forecast_year_min,
        top_n = top_n,
        hypoxia_threshold = hypoxia_threshold,
        material_share = material_share,
        plot_n = plot_n,
        stress_threshold = 5.0
    )

    return { 
        "path_source": "existing_scenario_paths",
        "region_codes_norm": region_codes_norm,
        "station_meta": station_meta,
        "station_keys": station_keys,
        "scenarios": forecast_scenarios,
        "future_year_min": forecast_year_min,
        "future_year_max": forecast_year_max,
        "future_station_year": future_station_year,
        "history_annual": history_annual,
        "history_plot": history_plot,
        "projection_plot": projection_plot,
        "projection_future_only": projection_future_only,
        "crossing_candidates": crossing_candidates,
        "hypoxia_candidates": hypoxia_candidates,
        "stress_candidates": stress_candidates,
        "crossing_display": crossing_display,
        "hypoxia_display": hypoxia_display,
        "stress_display": stress_display,
        "plot_studies": plot_studies
    }
```

## Tutorial Phase 5: Final Insights And Recommendations

Key findings:
- Station baseline behavior can be reduced into interpretable water-state regimes.
- Stable stations and regime-changing stations behave differently enough to justify a holdout split.
- The stage 6 model captures broad water-temperature response better than extreme events.
- Warm-season oxygen summaries are more decision-relevant than annual oxygen means.
- Forecast outputs should be read as regime-risk screening, not precise station-level prediction.

Data quality notes:
- Station records vary strongly in length and seasonal completeness.
- Some stations have recurring seasonal gaps that can mimic environmental transitions.
- Nutrient data remain too intermittent for this compact modeling path.

Recommendations:
- Treat projected regime crossing and oxygen stress as prioritization signals.
- Add river flow, tidal exchange, and local bathymetry if available.
- Use the NEX-GDDP-CMIP6 side-script outputs when they are available, and keep the demo ramps only as a lightweight workflow check.
